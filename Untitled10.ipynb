# 1. Data Understanding and Preparation

## 1.1 Problem framing

Network telemetry is captured per **bidirectional flow** (a conversation between two endpoints).
Each row is one flow described by 78 statistical features (durations, packet/byte counts, inter-arrival
times, TCP flag counts, header sizes, active/idle periods) plus a `Label` column that is either
`BENIGN` or the name of an attack family (DoS, DDoS, PortScan, brute-force/Patator, Bot,
Infiltration, Web Attack …).

**ML task.** Supervised **binary classification**: `is_attack = 0` for `BENIGN`, `1` otherwise.
The attack family is retained as a secondary label so we can report *per-family* recall — an
aggregate score can look excellent while a whole attack family is being missed.

**Why binary first?** XYZ's product decision is binary ("raise an alert / let the traffic pass").
Binary framing also pools the rare families, so the model can learn "not-normal" behaviour instead
of starving on 40 examples of a specific family. Attack-family attribution is treated as a second,
downstream problem (see §5).

**Business context that drives every later choice**

* **Extreme class imbalance** — benign traffic dominates. Accuracy is therefore useless
  (predicting "benign" for everything already scores ~80%+). We optimise **PR-AUC / recall at an
  acceptable false-positive rate**.
* **Asymmetric costs** — a missed attack (false negative) can mean a breach; a false positive
  costs analyst time and may block a paying customer. We encode this asymmetry explicitly as
  `FN_COST` vs `FP_COST` and *tune the decision threshold*, rather than blindly using 0.5.
* **Non-stationarity** — attacker behaviour changes. We test generalisation to an attack family
  never seen in training (§5.3) and to distribution monitoring in production (§5.5).

## 1.2 Success criteria agreed with the client

| Criterion | Target |
|---|---|
| Recall on attack traffic | ≥ 0.95 |
| False-positive rate on benign traffic | ≤ 1% |
| PR-AUC (attack = positive class) | ≥ 0.95 |
| Inference latency | ≤ 1 ms per flow (single-threaded, batchable) |
| Explainability | Per-alert drivers available to the SOC analyst |

```python
# =====================================================================================
# 0.2 CIC-IDS2017 schema + reproducible fallback data generator
#
# The assignment CSVs follow the CIC-IDS2017 "machine learning" schema: 78 flow features
# and a Label column (79 columns), with the raw header carrying stray leading spaces.
# If the CSVs are not present, we synthesise data with the *same schema, dtypes, class
# imbalance and data-quality defects* so the whole notebook remains executable.
# =====================================================================================
FEATURE_SCHEMA: tuple[str, ...] = (
    "Destination Port", "Flow Duration", "Total Fwd Packets", "Total Backward Packets",
    "Total Length of Fwd Packets", "Total Length of Bwd Packets", "Fwd Packet Length Max",
    "Fwd Packet Length Min", "Fwd Packet Length Mean", "Fwd Packet Length Std",
    "Bwd Packet Length Max", "Bwd Packet Length Min", "Bwd Packet Length Mean",
    "Bwd Packet Length Std", "Flow Bytes/s", "Flow Packets/s", "Flow IAT Mean", "Flow IAT Std",
    "Flow IAT Max", "Flow IAT Min", "Fwd IAT Total", "Fwd IAT Mean", "Fwd IAT Std", "Fwd IAT Max",
    "Fwd IAT Min", "Bwd IAT Total", "Bwd IAT Mean", "Bwd IAT Std", "Bwd IAT Max", "Bwd IAT Min",
    "Fwd PSH Flags", "Bwd PSH Flags", "Fwd URG Flags", "Bwd URG Flags", "Fwd Header Length",
    "Bwd Header Length", "Fwd Packets/s", "Bwd Packets/s", "Min Packet Length",
    "Max Packet Length", "Packet Length Mean", "Packet Length Std", "Packet Length Variance",
    "FIN Flag Count", "SYN Flag Count", "RST Flag Count", "PSH Flag Count", "ACK Flag Count",
    "URG Flag Count", "CWE Flag Count", "ECE Flag Count", "Down/Up Ratio", "Average Packet Size",
    "Avg Fwd Segment Size", "Avg Bwd Segment Size", "Fwd Header Length.1", "Fwd Avg Bytes/Bulk",
    "Fwd Avg Packets/Bulk", "Fwd Avg Bulk Rate", "Bwd Avg Bytes/Bulk", "Bwd Avg Packets/Bulk",
    "Bwd Avg Bulk Rate", "Subflow Fwd Packets", "Subflow Fwd Bytes", "Subflow Bwd Packets",
    "Subflow Bwd Bytes", "Init_Win_bytes_forward", "Init_Win_bytes_backward", "act_data_pkt_fwd",
    "min_seg_size_forward", "Active Mean", "Active Std", "Active Max", "Active Min", "Idle Mean",
    "Idle Std", "Idle Max", "Idle Min", "Label",
)

# Columns whose raw header keeps a leading blank (mirrors the real files, exercises cleaning).
_NO_LEADING_SPACE = {"Destination Port", "Fwd IAT Total", "Init_Win_bytes_forward",
                     "act_data_pkt_fwd", "min_seg_size_forward"}
RAW_SCHEMA: tuple[str, ...] = tuple(
    c if c in _NO_LEADING_SPACE else f" {c}" for c in FEATURE_SCHEMA
)

# Per-family behavioural profiles: weight, destination ports, log-normal (mu, sigma) for
# duration/packet-counts/packet-sizes, TCP flag probabilities, burstiness and idle rate.
_PROFILES: dict[str, dict[str, Any]] = {
    "BENIGN": dict(w=0.700, ports=(80, 443, 53, 389, 88, 445, 22, 3268, 8080, 123),
                   dur=(10.4, 2.5), fpk=(1.5, 1.0), bpk=(1.3, 1.1), fsz=(3.6, 1.6), bsz=(4.4, 1.7),
                   flags=dict(fin=0.22, syn=0.10, rst=0.03, psh=0.38, ack=0.72, urg=0.05),
                   burst=0.35, idle=0.30),
    "DoS Hulk": dict(w=0.085, ports=(80,), dur=(8.3, 1.1), fpk=(2.3, 0.5), bpk=(2.1, 0.6),
                     fsz=(5.3, 0.5), bsz=(6.4, 0.6),
                     flags=dict(fin=0.05, syn=0.02, rst=0.06, psh=0.95, ack=0.98, urg=0.02),
                     burst=0.05, idle=0.02),
    "DDoS": dict(w=0.060, ports=(80,), dur=(9.6, 1.5), fpk=(1.2, 0.5), bpk=(0.7, 0.6),
                 fsz=(6.7, 0.4), bsz=(3.0, 1.3),
                 flags=dict(fin=0.30, syn=0.05, rst=0.02, psh=0.70, ack=0.90, urg=0.02),
                 burst=0.10, idle=0.03),
    "PortScan": dict(w=0.055, ports="scan", dur=(3.1, 1.3), fpk=(0.05, 0.25), bpk=(0.02, 0.25),
                     fsz=(0.2, 0.4), bsz=(0.2, 0.5),
                     flags=dict(fin=0.02, syn=0.97, rst=0.55, psh=0.02, ack=0.10, urg=0.01),
                     burst=0.02, idle=0.01),
    "DoS GoldenEye": dict(w=0.022, ports=(80,), dur=(11.8, 1.0), fpk=(1.9, 0.6), bpk=(1.7, 0.7),
                          fsz=(5.0, 0.7), bsz=(6.0, 0.8),
                          flags=dict(fin=0.10, syn=0.04, rst=0.10, psh=0.85, ack=0.95, urg=0.03),
                          burst=0.60, idle=0.25),
    "FTP-Patator": dict(w=0.019, ports=(21,), dur=(12.6, 1.2), fpk=(1.7, 0.5), bpk=(1.9, 0.5),
                        fsz=(2.6, 0.6), bsz=(3.6, 0.7),
                        flags=dict(fin=0.35, syn=0.06, rst=0.05, psh=0.80, ack=0.95, urg=0.02),
                        burst=0.80, idle=0.55),
    "SSH-Patator": dict(w=0.014, ports=(22,), dur=(13.1, 1.1), fpk=(2.0, 0.5), bpk=(2.1, 0.5),
                        fsz=(4.3, 0.6), bsz=(4.6, 0.7),
                        flags=dict(fin=0.30, syn=0.05, rst=0.04, psh=0.85, ack=0.96, urg=0.02),
                        burst=0.75, idle=0.50),
    "DoS slowloris": dict(w=0.013, ports=(80,), dur=(14.4, 0.9), fpk=(1.3, 0.6), bpk=(1.0, 0.7),
                          fsz=(3.9, 0.7), bsz=(4.0, 0.9),
                          flags=dict(fin=0.15, syn=0.05, rst=0.20, psh=0.55, ack=0.90, urg=0.02),
                          burst=1.00, idle=0.85),
    "Web Attack - Brute Force": dict(w=0.014, ports=(80, 443), dur=(11.2, 2.0), fpk=(1.9, 0.9),
                                     bpk=(1.8, 1.0), fsz=(4.6, 1.3), bsz=(5.6, 1.4),
                                     flags=dict(fin=0.25, syn=0.09, rst=0.04, psh=0.55,
                                                ack=0.80, urg=0.04),
                                     burst=0.40, idle=0.30),
    "Bot": dict(w=0.011, ports=(8080, 443, 80), dur=(12.2, 1.8), fpk=(1.1, 0.7), bpk=(1.0, 0.8),
                fsz=(3.2, 1.0), bsz=(3.4, 1.2),
                flags=dict(fin=0.30, syn=0.12, rst=0.05, psh=0.45, ack=0.75, urg=0.03),
                burst=0.55, idle=0.60),
    "Infiltration": dict(w=0.007, ports=(445, 80, 443, 53), dur=(10.6, 2.4), fpk=(1.5, 1.0),
                         bpk=(1.4, 1.1), fsz=(3.8, 1.6), bsz=(4.5, 1.7),
                         flags=dict(fin=0.22, syn=0.11, rst=0.03, psh=0.40, ack=0.72, urg=0.05),
                         burst=0.36, idle=0.32),
}


def _simulate_family(label: str, n: int, rng: np.random.Generator) -> pd.DataFrame:
    """Generate ``n`` internally consistent CIC-IDS2017-style flows for one traffic family."""
    p = _PROFILES[label]
    zeros_i = np.zeros(n, dtype=np.int32)

    # --- primitive attributes ---------------------------------------------------------
    duration = np.clip(rng.lognormal(p["dur"][0], p["dur"][1], n), 1.0, 1.2e8)  # microseconds
    fwd_pkts = 1 + rng.poisson(np.clip(rng.lognormal(p["fpk"][0], p["fpk"][1], n), 0.0, 3000.0))
    bwd_pkts = rng.poisson(np.clip(rng.lognormal(p["bpk"][0], p["bpk"][1], n), 0.0, 3000.0))
    fwd_mean = np.clip(rng.lognormal(p["fsz"][0], p["fsz"][1], n), 0.0, 1460.0)
    bwd_mean = np.where(bwd_pkts > 0,
                        np.clip(rng.lognormal(p["bsz"][0], p["bsz"][1], n), 0.0, 1460.0), 0.0)

    fwd_bytes = np.round(fwd_pkts * fwd_mean)
    bwd_bytes = np.round(bwd_pkts * bwd_mean)
    total_pkts = fwd_pkts + bwd_pkts
    total_bytes = fwd_bytes + bwd_bytes
    dur_s = duration / 1e6

    # --- packet-length statistics (kept mutually consistent: min <= mean <= max) -------
    spread = rng.uniform(0.05, 0.95, n)
    fwd_max = np.minimum(1460.0, fwd_mean * (1.0 + spread))
    fwd_min = np.where(fwd_pkts > 1, fwd_mean * rng.uniform(0.0, 0.6, n), fwd_mean)
    fwd_std = np.where(fwd_pkts > 1, (fwd_max - fwd_min) * rng.uniform(0.2, 0.6, n), 0.0)
    bwd_max = np.minimum(1460.0, bwd_mean * (1.0 + spread))
    bwd_min = np.where(bwd_pkts > 1, bwd_mean * rng.uniform(0.0, 0.6, n), bwd_mean)
    bwd_std = np.where(bwd_pkts > 1, (bwd_max - bwd_min) * rng.uniform(0.2, 0.6, n), 0.0)
    pkt_min = np.where(bwd_pkts > 0, np.minimum(fwd_min, bwd_min), fwd_min)
    pkt_max = np.maximum(fwd_max, bwd_max)
    pkt_mean = total_bytes / np.maximum(total_pkts, 1)
    pkt_std = (pkt_max - pkt_min) * rng.uniform(0.15, 0.55, n)

    # --- inter-arrival times ----------------------------------------------------------
    iat_mean = duration / np.maximum(total_pkts - 1, 1)
    iat_std = iat_mean * rng.uniform(0.05, 1.0 + 3.0 * p["burst"], n)
    iat_max = iat_mean + iat_std * rng.uniform(1.0, 3.0, n)
    iat_min = iat_mean * rng.uniform(0.0, 0.4, n)
    f_iat_tot = duration * rng.uniform(0.55, 1.0, n)
    f_iat_mean = f_iat_tot / np.maximum(fwd_pkts - 1, 1)
    f_iat_std = f_iat_mean * rng.uniform(0.05, 1.5, n)
    f_iat_max = f_iat_mean + f_iat_std * rng.uniform(1.0, 3.0, n)
    f_iat_min = f_iat_mean * rng.uniform(0.0, 0.4, n)
    b_iat_tot = np.where(bwd_pkts > 1, duration * rng.uniform(0.30, 0.95, n), 0.0)
    b_iat_mean = b_iat_tot / np.maximum(bwd_pkts - 1, 1)
    b_iat_std = b_iat_mean * rng.uniform(0.05, 1.5, n)
    b_iat_max = b_iat_mean + b_iat_std * rng.uniform(1.0, 3.0, n)
    b_iat_min = b_iat_mean * rng.uniform(0.0, 0.4, n)

    # --- headers, flags, windows ------------------------------------------------------
    seg_fwd = rng.choice(np.array([20, 32], dtype=np.int32), n)
    seg_bwd = rng.choice(np.array([20, 32], dtype=np.int32), n)
    fwd_header = fwd_pkts * seg_fwd
    bwd_header = bwd_pkts * seg_bwd

    f = p["flags"]
    fin = (rng.random(n) < f["fin"]).astype(np.int8)
    syn = (rng.random(n) < f["syn"]).astype(np.int8)
    rst = (rng.random(n) < f["rst"]).astype(np.int8)
    psh = ((rng.random(n) < f["psh"]) * rng.integers(1, 3, n)).astype(np.int8)
    ack = (rng.random(n) < f["ack"]).astype(np.int8)
    urg = (rng.random(n) < f["urg"]).astype(np.int8)
    ece = (rst * (rng.random(n) < 0.30)).astype(np.int8)

    init_win_f = rng.choice(np.array([-1, 256, 8192, 29200, 65535], dtype=np.int32), n)
    init_win_b = np.where(bwd_pkts > 0,
                          rng.choice(np.array([-1, 235, 8192, 26883, 65535], dtype=np.int32), n),
                          -1).astype(np.int32)

    # --- active / idle periods --------------------------------------------------------
    has_idle = rng.random(n) < p["idle"]
    idle_mean = np.where(has_idle, duration * rng.uniform(0.30, 0.80, n), 0.0)
    idle_std = np.where(has_idle, idle_mean * rng.uniform(0.0, 0.50, n), 0.0)
    idle_max = np.where(has_idle, idle_mean * rng.uniform(1.0, 1.8, n), 0.0)
    idle_min = np.where(has_idle, idle_mean * rng.uniform(0.20, 1.0, n), 0.0)
    active_mean = np.where(has_idle,
                           np.maximum(duration - idle_mean, 0.0) * rng.uniform(0.4, 1.0, n), 0.0)
    active_std = active_mean * rng.uniform(0.0, 0.50, n)
    active_max = active_mean * rng.uniform(1.0, 1.6, n)
    active_min = active_mean * rng.uniform(0.20, 1.0, n)

    ports = (rng.integers(1, 65_535, n) if p["ports"] == "scan"
             else rng.choice(np.array(p["ports"], dtype=np.int32), n))

    data: dict[str, Any] = {
        "Destination Port": ports.astype(np.int32),
        "Flow Duration": duration.astype(np.int64),
        "Total Fwd Packets": fwd_pkts.astype(np.int32),
        "Total Backward Packets": bwd_pkts.astype(np.int32),
        "Total Length of Fwd Packets": fwd_bytes.astype(np.int64),
        "Total Length of Bwd Packets": bwd_bytes.astype(np.int64),
        "Fwd Packet Length Max": fwd_max, "Fwd Packet Length Min": fwd_min,
        "Fwd Packet Length Mean": fwd_mean, "Fwd Packet Length Std": fwd_std,
        "Bwd Packet Length Max": bwd_max, "Bwd Packet Length Min": bwd_min,
        "Bwd Packet Length Mean": bwd_mean, "Bwd Packet Length Std": bwd_std,
        "Flow Bytes/s": total_bytes / dur_s, "Flow Packets/s": total_pkts / dur_s,
        "Flow IAT Mean": iat_mean, "Flow IAT Std": iat_std, "Flow IAT Max": iat_max,
        "Flow IAT Min": iat_min,
        "Fwd IAT Total": f_iat_tot, "Fwd IAT Mean": f_iat_mean, "Fwd IAT Std": f_iat_std,
        "Fwd IAT Max": f_iat_max, "Fwd IAT Min": f_iat_min,
        "Bwd IAT Total": b_iat_tot, "Bwd IAT Mean": b_iat_mean, "Bwd IAT Std": b_iat_std,
        "Bwd IAT Max": b_iat_max, "Bwd IAT Min": b_iat_min,
        "Fwd PSH Flags": (psh > 0).astype(np.int8), "Bwd PSH Flags": np.zeros(n, np.int8),
        "Fwd URG Flags": urg, "Bwd URG Flags": np.zeros(n, np.int8),
        "Fwd Header Length": fwd_header.astype(np.int32),
        "Bwd Header Length": bwd_header.astype(np.int32),
        "Fwd Packets/s": fwd_pkts / dur_s, "Bwd Packets/s": bwd_pkts / dur_s,
        "Min Packet Length": pkt_min, "Max Packet Length": pkt_max,
        "Packet Length Mean": pkt_mean, "Packet Length Std": pkt_std,
        "Packet Length Variance": pkt_std ** 2,
        "FIN Flag Count": fin, "SYN Flag Count": syn, "RST Flag Count": rst,
        "PSH Flag Count": psh, "ACK Flag Count": ack, "URG Flag Count": urg,
        "CWE Flag Count": np.zeros(n, np.int8), "ECE Flag Count": ece,
        "Down/Up Ratio": np.floor(bwd_pkts / np.maximum(fwd_pkts, 1)).astype(np.int16),
        "Average Packet Size": pkt_mean,
        "Avg Fwd Segment Size": fwd_mean, "Avg Bwd Segment Size": bwd_mean,
        "Fwd Header Length.1": fwd_header.astype(np.int32),
        "Fwd Avg Bytes/Bulk": zeros_i, "Fwd Avg Packets/Bulk": zeros_i,
        "Fwd Avg Bulk Rate": zeros_i, "Bwd Avg Bytes/Bulk": zeros_i,
        "Bwd Avg Packets/Bulk": zeros_i, "Bwd Avg Bulk Rate": zeros_i,
        "Subflow Fwd Packets": fwd_pkts.astype(np.int32),
        "Subflow Fwd Bytes": fwd_bytes.astype(np.int64),
        "Subflow Bwd Packets": bwd_pkts.astype(np.int32),
        "Subflow Bwd Bytes": bwd_bytes.astype(np.int64),
        "Init_Win_bytes_forward": init_win_f, "Init_Win_bytes_backward": init_win_b,
        "act_data_pkt_fwd": np.maximum(fwd_pkts - 1, 0).astype(np.int32),
        "min_seg_size_forward": seg_fwd,
        "Active Mean": active_mean, "Active Std": active_std, "Active Max": active_max,
        "Active Min": active_min,
        "Idle Mean": idle_mean, "Idle Std": idle_std, "Idle Max": idle_max, "Idle Min": idle_min,
        "Label": np.repeat(label, n),
    }
    return pd.DataFrame(data)


def simulate_cicids2017(n_rows: int, seed: int) -> pd.DataFrame:
    """Build a reproducible CIC-IDS2017-schema dataset, defects included.

    Injected defects mirror the published dataset: infinite ``Flow Bytes/s`` on zero-duration
    flows, a few missing values, exact duplicate rows, all-zero "bulk" columns and ~0.6%
    annotation noise. They exist so the cleaning code in 1.4 is genuinely exercised.
    """
    rng = np.random.default_rng(seed)
    frames = [_simulate_family(lbl, max(60, int(round(n_rows * prof["w"]))), rng)
              for lbl, prof in _PROFILES.items()]
    df = pd.concat(frames, ignore_index=True)
    df = df.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    n = len(df)

    inf_idx = rng.choice(n, size=int(0.004 * n), replace=False)
    df.loc[inf_idx, "Flow Duration"] = 0
    df.loc[inf_idx, ["Flow Bytes/s", "Flow Packets/s"]] = np.inf
    nan_idx = rng.choice(n, size=int(0.002 * n), replace=False)
    df.loc[nan_idx, "Flow Bytes/s"] = np.nan

    benign_idx = df.index[df["Label"] == "BENIGN"]
    attack_idx = df.index[df["Label"] != "BENIGN"]
    df.loc[rng.choice(attack_idx, size=int(0.004 * len(attack_idx)), replace=False), "Label"] = "BENIGN"
    df.loc[rng.choice(benign_idx, size=int(0.002 * len(benign_idx)), replace=False), "Label"] = "DoS Hulk"

    dup = df.loc[rng.choice(n, size=int(0.015 * n), replace=False)]
    df = pd.concat([df, dup], ignore_index=True).sample(frac=1.0, random_state=seed + 1)
    df = df.reset_index(drop=True)

    df = df[list(FEATURE_SCHEMA)]
    df.columns = list(RAW_SCHEMA)   # reinstate the messy raw header
    return df


print(f"Schema defined: {len(FEATURE_SCHEMA)} columns "
      f"({len(FEATURE_SCHEMA) - 1} features + label), {len(_PROFILES)} traffic families.")
```

```python
# =====================================================================================
# 1.3 Load the data (real CSVs when available, simulator otherwise) and merge
# =====================================================================================
def find_dataset_files(cfg: Config = CFG) -> list[Path]:
    """Return the first directory's worth of assignment CSVs that actually exists on disk."""
    for directory in cfg.search_dirs:
        base = Path(directory)
        if not base.is_dir():
            continue
        hits = [base / name for name in cfg.data_files if (base / name).is_file()]
        if hits:
            return hits
    return []


def load_datasets(cfg: Config = CFG) -> tuple[pd.DataFrame, str]:
    """Load and vertically merge the source datasets.

    Returns
    -------
    (frame, mode) where ``mode`` is ``"real-csv"`` or ``"simulated"``.
    """
    paths = find_dataset_files(cfg)
    if paths:
        frames = [pd.read_csv(p, low_memory=False) for p in paths]
        for path, frame in zip(paths, frames):
            print(f"  loaded {path}  ->  {frame.shape[0]:,} rows x {frame.shape[1]} cols")
        merged = pd.concat(frames, ignore_index=True)
        return merged, "real-csv"

    print("  Dataset1.csv / Dataset2.csv / Dataset3.csv were not found in:")
    print("   ", ", ".join(cfg.search_dirs))
    print("  -> DATA MODE = SIMULATED: generating a CIC-IDS2017-schema dataset "
          f"(seed={cfg.random_state}).")
    print("     Upload the real CSVs and re-run this cell to switch to DATA MODE = real-csv.")
    return simulate_cicids2017(cfg.simulated_rows, cfg.random_state), "simulated"


def stratified_subsample(df: pd.DataFrame, n: int, seed: int, label_col: str) -> pd.DataFrame:
    """Down-sample to ~``n`` rows while preserving the class proportions."""
    if len(df) <= n:
        return df.reset_index(drop=True)
    frac = n / len(df)
    parts = [g.sample(n=max(1, int(round(len(g) * frac))), random_state=seed)
             for _, g in df.groupby(label_col, sort=False)]
    return (pd.concat(parts, ignore_index=True)
              .sample(frac=1.0, random_state=seed)
              .reset_index(drop=True))


raw_df, DATA_MODE = load_datasets()
RAW_LABEL_COL = next(c for c in raw_df.columns if c.strip().lower() == "label")
print(f"\nMerged shape        : {raw_df.shape[0]:,} rows x {raw_df.shape[1]} columns")

raw_df = stratified_subsample(raw_df, CFG.sample_size, CFG.random_state, RAW_LABEL_COL)
print(f"After stratified cap: {raw_df.shape[0]:,} rows  (cap = {CFG.sample_size:,})")
print(f"DATA MODE           : {DATA_MODE}")
raw_df.head()
```

## 1.4 Cleaning strategy and *why* each step is applied

| Issue in the raw flow data | Consequence if ignored | Treatment |
|---|---|---|
| Header names carry stray spaces (`" Flow Duration"`) | Silent `KeyError`s, duplicated logic | `str.strip()` normalisation |
| `Flow Bytes/s`, `Flow Packets/s` become `±inf` when `Flow Duration == 0` | `sklearn` raises, or scaling explodes | replace with `NaN`, then median-impute **inside the pipeline** |
| A few genuine `NaN`s | rows silently dropped by some estimators | imputed inside the CV pipeline (no leakage from test data) |
| Exact duplicate rows (sensor re-emission / merged files overlapping) | over-optimistic scores, duplicated rows across the train/test split | `drop_duplicates()` **before** splitting |
| Zero-variance columns (`Bwd PSH Flags`, all `*Bulk*` columns, `CWE Flag Count`) | no information, wasted compute, unstable importances | dropped |
| Exactly duplicated columns (`Fwd Header Length.1` == `Fwd Header Length`) | double-counted signal, misleading importances | dropped |
| Label spelling variants / mojibake (`Web Attack \x96 Brute Force`) | family split across several classes | canonicalised to ASCII |
| `float64` everywhere | ~2× RAM on a Colab instance | down-cast to `float32` |

Two deliberate decisions worth defending in the report:

* **Imputation and scaling live inside the `Pipeline`, not in this cleaning step.** Fitting an
  imputer or scaler on the whole dataset leaks test-set statistics into training and inflates
  cross-validation scores. Only *row/column removal* and *dtype* work happens here.
* **Negative values are kept.** `Init_Win_bytes_backward = -1` is the sentinel for "no backward
  window", i.e. real signal, not an error.

```python
# =====================================================================================
# 1.4 Cleaning
# =====================================================================================
def canonicalise_labels(series: pd.Series, benign_token: str = CFG.benign_token) -> pd.Series:
    """Normalise label text: strip, fix mojibake dashes, collapse inner whitespace."""
    out = (series.astype(str)
                 .str.replace("\x96", "-", regex=False)
                 .str.replace("\u2013", "-", regex=False)
                 .str.replace("\u2014", "-", regex=False)
                 .str.strip()
                 .str.replace(r"\s+", " ", regex=True))
    return out.mask(out.str.upper() == benign_token, benign_token)


def clean_dataset(df: pd.DataFrame, label_col: str = CFG.label_col) -> tuple[pd.DataFrame, dict]:
    """Apply the cleaning contract described above.

    Returns the cleaned frame and a report dict used for the audit trail (§7).
    """
    report: dict[str, Any] = {"rows_in": int(len(df)), "cols_in": int(df.shape[1])}
    out = df.copy()
    out.columns = [str(c).replace("\ufeff", "").strip() for c in out.columns]

    label = canonicalise_labels(out[label_col])
    features = out.drop(columns=[label_col]).apply(pd.to_numeric, errors="coerce")

    n_inf = int(np.isinf(features.to_numpy(dtype="float64", na_value=np.nan)).sum())
    features = features.replace([np.inf, -np.inf], np.nan)
    report["inf_values_replaced"] = n_inf

    out = pd.concat([features, label.rename(label_col)], axis=1)

    before = len(out)
    out = out.drop_duplicates().reset_index(drop=True)
    report["duplicate_rows_dropped"] = int(before - len(out))

    bad_label = out[label_col].str.lower().isin(["", "nan", "none", "na"])
    out = out[~bad_label].reset_index(drop=True)
    report["rows_missing_label_dropped"] = int(before - report["duplicate_rows_dropped"] - len(out))

    feats = out.drop(columns=[label_col])
    constant = [c for c in feats.columns if feats[c].nunique(dropna=False) <= 1]
    feats = feats.drop(columns=constant)
    report["constant_columns_dropped"] = constant

    duplicated_cols: list[str] = []
    seen: dict[str, str] = {}
    for col in feats.columns:
        key = joblib.hash(np.ascontiguousarray(feats[col].to_numpy(dtype="float64", na_value=-9.99e17)))
        if key in seen:
            duplicated_cols.append(col)
        else:
            seen[key] = col
    feats = feats.drop(columns=duplicated_cols)
    report["duplicate_columns_dropped"] = duplicated_cols

    float_cols = feats.select_dtypes(include=["float64"]).columns
    feats[float_cols] = feats[float_cols].astype(np.float32)

    out = pd.concat([feats, out[label_col]], axis=1)
    report["rows_out"] = int(len(out))
    report["cols_out"] = int(out.shape[1])
    report["missing_cells_remaining"] = int(feats.isna().sum().sum())
    return out, report


clean_df, CLEAN_REPORT = clean_dataset(raw_df)

print("Cleaning report")
print("-" * 74)
for key, value in CLEAN_REPORT.items():
    shown = value if not isinstance(value, list) else (value if len(value) <= 8 else f"{value[:8]} (+{len(value) - 8} more)")
    print(f"{key:<28}: {shown}")
print("-" * 74)
print(f"Removed {CLEAN_REPORT['cols_in'] - CLEAN_REPORT['cols_out']} columns and "
      f"{CLEAN_REPORT['rows_in'] - CLEAN_REPORT['rows_out']:,} rows.")
print(f"Memory now: {clean_df.memory_usage(deep=True).sum() / 1024**2:,.1f} MB")
clean_df.iloc[:, :10].describe().T
```

## 1.5 Feature engineering

Raw CICFlowMeter features are *absolute* counters. Attacks are much better described by
**rates, ratios and asymmetry** — a 5 MB download and a 5 MB flood differ in *how* the bytes were
delivered, not in the total. Every engineered feature below encodes an explicit attack hypothesis:

| Engineered feature(s) | Attack hypothesis encoded |
|---|---|
| `fe_bytes_per_second`, `fe_packets_per_second`, `fe_bytes_per_packet` | Volumetric floods (Hulk/DDoS) push extreme rates; slow-rate DoS pushes extremely low ones |
| `fe_fwd_bwd_packet_ratio`, `fe_fwd_bwd_byte_ratio`, `fe_bwd_share_of_bytes` | Legitimate sessions are conversational; scans and floods are one-directional |
| `fe_is_one_way_flow`, `fe_no_response` | No server reply at all = probe or spoofed source |
| `fe_flag_rate`, `fe_syn_without_ack`, `fe_rst_rate` | Half-open SYN scans and RST storms never complete the handshake |
| `fe_iat_burstiness`, `fe_iat_range_ratio` | Automated tooling is metronomic (low dispersion); humans are bursty |
| `fe_idle_active_ratio`, `fe_idle_share` | Slowloris/Patator hold connections open and idle |
| `fe_header_overhead`, `fe_payload_per_packet` | Scans are header-only; exfiltration is payload-heavy |
| `fe_port_class`, `fe_is_web_port`, `fe_is_admin_port`, `fe_is_ephemeral_port` | Service context: brute force targets 21/22, scans hit high random ports |
| `fe_pkt_len_cv`, `fe_fwd_len_cv` | Machine-generated packets are uniform in size |
| `fe_log_*` | Compress 8-order-of-magnitude heavy tails so linear/distance models are usable |
| `fe_bytes_per_second_x_oneway` | Interaction: high rate **and** unanswered is the DDoS signature |

Design rules followed: **no target leakage** (nothing is derived from `Label`), **row-local
computation only** (every feature depends on that flow alone, so it is computable at inference
time), and **division guarded** with `+1`/`eps` denominators so no `inf` is reintroduced.

```python
# =====================================================================================
# 1.5 Feature engineering
# =====================================================================================
EPS = 1e-6


def _col(df: pd.DataFrame, name: str, default: float = 0.0) -> pd.Series:
    """Fetch a column, or a constant fallback if cleaning removed it (keeps FE robust)."""
    if name in df.columns:
        return pd.to_numeric(df[name], errors="coerce").astype("float64")
    return pd.Series(default, index=df.index, dtype="float64")


def engineer_features(df: pd.DataFrame, label_col: str = CFG.label_col) -> pd.DataFrame:
    """Add rate/ratio/asymmetry/context features. Pure function: input is never mutated."""
    out = df.copy()

    fwd_pk, bwd_pk = _col(out, "Total Fwd Packets"), _col(out, "Total Backward Packets")
    fwd_by, bwd_by = _col(out, "Total Length of Fwd Packets"), _col(out, "Total Length of Bwd Packets")
    duration_us = _col(out, "Flow Duration")
    dur_s = duration_us / 1e6
    total_pk = fwd_pk + bwd_pk
    total_by = fwd_by + bwd_by
    port = _col(out, "Destination Port")

    # --- volume & rate ----------------------------------------------------------------
    out["fe_total_packets"] = total_pk
    out["fe_total_bytes"] = total_by
    out["fe_bytes_per_second"] = total_by / (dur_s + EPS)
    out["fe_packets_per_second"] = total_pk / (dur_s + EPS)
    out["fe_bytes_per_packet"] = total_by / (total_pk + EPS)

    # --- directional asymmetry --------------------------------------------------------
    out["fe_fwd_bwd_packet_ratio"] = fwd_pk / (bwd_pk + 1.0)
    out["fe_fwd_bwd_byte_ratio"] = fwd_by / (bwd_by + 1.0)
    out["fe_bwd_share_of_bytes"] = bwd_by / (total_by + EPS)
    out["fe_is_one_way_flow"] = (bwd_pk == 0).astype("int8")
    out["fe_no_response"] = ((bwd_pk == 0) & (fwd_pk > 0)).astype("int8")

    # --- TCP control-flag behaviour ---------------------------------------------------
    flag_cols = ["FIN Flag Count", "SYN Flag Count", "RST Flag Count", "PSH Flag Count",
                 "ACK Flag Count", "URG Flag Count", "ECE Flag Count"]
    flag_total = sum(_col(out, c) for c in flag_cols)
    syn, ack, rst = _col(out, "SYN Flag Count"), _col(out, "ACK Flag Count"), _col(out, "RST Flag Count")
    out["fe_flag_total"] = flag_total
    out["fe_flag_rate"] = flag_total / (total_pk + EPS)
    out["fe_syn_without_ack"] = ((syn > 0) & (ack == 0)).astype("int8")
    out["fe_rst_rate"] = rst / (total_pk + EPS)

    # --- timing regularity ------------------------------------------------------------
    iat_mean, iat_std = _col(out, "Flow IAT Mean"), _col(out, "Flow IAT Std")
    iat_max, iat_min = _col(out, "Flow IAT Max"), _col(out, "Flow IAT Min")
    out["fe_iat_burstiness"] = iat_std / (iat_mean + EPS)          # coefficient of variation
    out["fe_iat_range_ratio"] = (iat_max - iat_min) / (iat_mean + EPS)
    active_mean, idle_mean = _col(out, "Active Mean"), _col(out, "Idle Mean")
    out["fe_idle_active_ratio"] = idle_mean / (active_mean + EPS)
    out["fe_idle_share"] = idle_mean / (duration_us + EPS)

    # --- packet-size regularity -------------------------------------------------------
    pkt_mean, pkt_std = _col(out, "Packet Length Mean"), _col(out, "Packet Length Std")
    out["fe_pkt_len_cv"] = pkt_std / (pkt_mean + EPS)
    out["fe_fwd_len_cv"] = _col(out, "Fwd Packet Length Std") / (_col(out, "Fwd Packet Length Mean") + EPS)

    # --- protocol overhead ------------------------------------------------------------
    header_total = _col(out, "Fwd Header Length") + _col(out, "Bwd Header Length")
    out["fe_header_overhead"] = header_total / (total_by + EPS)
    out["fe_payload_per_packet"] = (total_by - header_total).clip(lower=0) / (total_pk + EPS)

    # --- service context --------------------------------------------------------------
    out["fe_port_class"] = np.select(
        [port <= 1023, port <= 49151], [0, 1], default=2).astype("int8")  # system / user / dynamic
    out["fe_is_web_port"] = port.isin([80, 443, 8080, 8443]).astype("int8")
    out["fe_is_admin_port"] = port.isin([21, 22, 23, 3389, 445, 139]).astype("int8")
    out["fe_is_ephemeral_port"] = (port >= 49152).astype("int8")

    # --- heavy-tail compression (helps linear / distance-based learners) --------------
    for src, name in [(duration_us, "fe_log_duration"), (total_by, "fe_log_bytes"),
                      (total_pk, "fe_log_packets"), (out["fe_bytes_per_second"], "fe_log_bytes_per_s"),
                      (out["fe_packets_per_second"], "fe_log_packets_per_s")]:
        out[name] = np.log1p(pd.Series(src, index=out.index).clip(lower=0))

    # --- interaction: fast AND unanswered = flood signature ---------------------------
    out["fe_bytes_per_second_x_oneway"] = out["fe_log_bytes_per_s"] * out["fe_is_one_way_flow"]

    # guard: engineered maths must not reintroduce inf / absurd magnitudes
    new_cols = [c for c in out.columns if c.startswith("fe_")]
    out[new_cols] = (out[new_cols].replace([np.inf, -np.inf], np.nan)
                                  .clip(lower=-1e12, upper=1e12)
                                  .astype(np.float32))
    return out


feature_df = engineer_features(clean_df)
NEW_FEATURES = [c for c in feature_df.columns if c.startswith("fe_")]
print(f"Engineered {len(NEW_FEATURES)} new features "
      f"({clean_df.shape[1] - 1} raw -> {feature_df.shape[1] - 1} total).")
print(f"inf values introduced: "
      f"{int(np.isinf(feature_df[NEW_FEATURES].to_numpy(dtype='float64', na_value=0.0)).sum())}")
display(feature_df[NEW_FEATURES].describe().T[["mean", "std", "min", "50%", "max"]])

# quick sanity check that the new features carry signal
_y = (feature_df[CFG.label_col].str.upper() != CFG.benign_token).astype(int)
signal = (feature_df[NEW_FEATURES]
          .apply(lambda s: s.fillna(s.median()).corr(_y))
          .abs().sort_values(ascending=False))
print("\nTop 10 engineered features by |correlation| with the target:")
display(signal.head(10).to_frame("abs_corr_with_target"))
```

```python
# =====================================================================================
# 1.6 Target definition, train/test split, and correlation pruning
#
# Split FIRST, then prune correlated features using training data only: any decision that
# looks at the test set (even an unsupervised one) is a leakage risk in a security model
# that must generalise to future traffic.
# =====================================================================================
y_family = feature_df[CFG.label_col].astype(str)                       # multi-class, for reporting
y = (y_family.str.upper() != CFG.benign_token).astype(int).rename("is_attack")
X = feature_df.drop(columns=[CFG.label_col])

X_train, X_test, y_train, y_test, fam_train, fam_test = train_test_split(
    X, y, y_family, test_size=CFG.test_size, stratify=y, random_state=CFG.random_state
)


def correlated_columns(frame: pd.DataFrame, threshold: float) -> list[str]:
    """Greedily list columns whose |Pearson r| with an earlier column exceeds ``threshold``."""
    corr = frame.corr(numeric_only=True).abs()
    upper = corr.where(np.triu(np.ones(corr.shape, dtype=bool), k=1))
    return [c for c in upper.columns if (upper[c] > threshold).any()]


redundant = correlated_columns(X_train, CFG.corr_threshold)
X_train = X_train.drop(columns=redundant)
X_test = X_test.drop(columns=redundant)
FEATURES: list[str] = list(X_train.columns)

print(f"Dropped {len(redundant)} features with |r| > {CFG.corr_threshold} "
      f"(e.g. {redundant[:6]}{' ...' if len(redundant) > 6 else ''})")
print(f"Final feature count : {len(FEATURES)} "
      f"({sum(c.startswith('fe_') for c in FEATURES)} engineered, "
      f"{sum(not c.startswith('fe_') for c in FEATURES)} raw)")
print(f"Train / test        : {X_train.shape[0]:,} / {X_test.shape[0]:,} flows")
print(f"Attack rate         : train {y_train.mean():.3%} | test {y_test.mean():.3%} "
      "(stratified, so both match the population)")
print(f"Attack flows        : train {int(y_train.sum()):,} | test {int(y_test.sum()):,}")

top_signal = (X_train.apply(lambda s: s.fillna(s.median()).corr(y_train)).abs()
                     .sort_values(ascending=False).head(18).index)
plt.figure(figsize=(11, 8.5))
sns.heatmap(X_train[top_signal].corr(), cmap="coolwarm", center=0, annot=False,
            square=True, cbar_kws={"shrink": 0.75, "label": "Pearson r"})
plt.title("Correlation among the 18 most target-relevant surviving features")
plt.show()
```

# 2. Model Selection

## 2.1 Selection protocol

Choosing a model on a single train/test score would be a coin flip on this data, so the protocol is:

1. **Two dumb baselines first.** A majority-class predictor and a stratified random predictor set
   the floor. Any candidate that does not clearly beat them adds nothing.
2. **Five candidate families** spanning the bias/variance and interpretability spectrum
   (linear, distance-based, single tree, bagged trees, boosted trees).
3. **Identical `Pipeline` for every candidate** (median imputation → optional scaling → estimator),
   so preprocessing is refit inside every fold. Scaling is applied only to the models that need it
   (logistic regression, kNN); trees are scale-invariant.
4. **Stratified 5-fold cross-validation** on a stratified subsample of the *training* set — the test
   set is untouched until §3.
5. **Imbalance-aware ranking metric: PR-AUC (average precision).** ROC-AUC is optimistic under
   heavy imbalance because the huge benign class makes the false-positive rate look tiny;
   precision-recall focuses on the rare positive class we care about. `recall`, `precision`,
   `f1` and `balanced accuracy` are reported alongside for context.
6. **Class imbalance handled by re-weighting** (`class_weight="balanced"`) rather than by SMOTE
   over-sampling: re-weighting does not synthesise fictitious network flows (a synthetic flow can be
   physically impossible, e.g. more bytes than packets allow) and it costs nothing at inference.

### Candidate rationale

| Candidate | Why it is in the shortlist | Expected weakness |
|---|---|---|
| Majority / stratified dummy | Sanity floor; exposes the accuracy trap | No signal at all |
| Logistic Regression | Fast, calibrated, coefficients are auditable — a good "explainable fallback" for the SOC | Linear boundary cannot express `high rate AND no reply` interactions |
| k-Nearest Neighbours | Non-parametric; attacks form tight behavioural clusters | Memory-bound and slow at inference (kills the 1 ms latency target) |
| Decision Tree | Human-readable rules a security analyst can validate | High variance; overfits noisy flow counters |
| Random Forest | Bagged trees handle mixed scales, skew, interactions and outliers; `balanced_subsample` targets imbalance directly | Larger memory footprint |
| Hist Gradient Boosting | Usually the strongest tabular learner; native handling of missing values; very fast on binned data | More hyperparameters, slightly harder to explain |

```python
# =====================================================================================
# 2.2 Uniform pipeline factory + candidate shortlist
# =====================================================================================
def build_pipeline(estimator, scale: bool = False) -> Pipeline:
    """Wrap an estimator in the shared preprocessing pipeline.

    Median imputation is fitted *inside* each CV fold, so no test-fold statistic ever
    reaches the training data.
    """
    steps: list[tuple[str, Any]] = [("imputer", SimpleImputer(strategy="median"))]
    if scale:
        steps.append(("scaler", StandardScaler()))
    steps.append(("clf", estimator))
    return Pipeline(steps)


CANDIDATES: dict[str, tuple[Any, bool]] = {
    "Dummy (majority)": (DummyClassifier(strategy="most_frequent"), False),
    "Dummy (stratified)": (DummyClassifier(strategy="stratified", random_state=CFG.random_state), False),
    "Logistic Regression": (
        LogisticRegression(max_iter=3000, class_weight="balanced", random_state=CFG.random_state),
        True,
    ),
    "k-Nearest Neighbours": (KNeighborsClassifier(n_neighbors=7, n_jobs=-1), True),
    "Decision Tree": (
        DecisionTreeClassifier(class_weight="balanced", min_samples_leaf=5,
                               random_state=CFG.random_state),
        False,
    ),
    "Random Forest": (
        RandomForestClassifier(n_estimators=200, min_samples_leaf=1, n_jobs=-1,
                               class_weight="balanced_subsample", random_state=CFG.random_state),
        False,
    ),
    "Hist Gradient Boosting": (
        HistGradientBoostingClassifier(class_weight="balanced", random_state=CFG.random_state),
        False,
    ),
}

SCORING = {
    "pr_auc": "average_precision",
    "roc_auc": "roc_auc",
    "f1": "f1",
    "recall": "recall",
    "precision": "precision",
    "balanced_acc": "balanced_accuracy",
}

# Stratified CV subsample: keeps 5-fold x 7 candidates inside a Colab CPU budget.
if len(X_train) > CFG.cv_sample_size:
    X_cv, _, y_cv, _ = train_test_split(
        X_train, y_train, train_size=CFG.cv_sample_size,
        stratify=y_train, random_state=CFG.random_state
    )
else:
    X_cv, y_cv = X_train, y_train

CV = StratifiedKFold(n_splits=CFG.cv_folds, shuffle=True, random_state=CFG.random_state)
print(f"Cross-validation set: {len(X_cv):,} flows | attack rate {y_cv.mean():.3%} | "
      f"{CFG.cv_folds}-fold stratified | ranking metric = PR-AUC")
```

```python
# =====================================================================================
# 2.3 Cross-validated comparison of all candidates
# =====================================================================================
rows: list[dict[str, Any]] = []
for name, (estimator, needs_scaling) in CANDIDATES.items():
    started = time.perf_counter()
    scores = cross_validate(
        build_pipeline(estimator, needs_scaling), X_cv, y_cv,
        cv=CV, scoring=SCORING, return_train_score=True, error_score="raise",
    )
    row: dict[str, Any] = {"model": name}
    for metric in SCORING:
        row[f"{metric}"] = scores[f"test_{metric}"].mean()
        row[f"{metric}_sd"] = scores[f"test_{metric}"].std()
    row["train_pr_auc"] = scores["train_pr_auc"].mean()
    row["overfit_gap"] = row["train_pr_auc"] - row["pr_auc"]
    row["fit_s"] = scores["fit_time"].mean()
    row["score_s"] = scores["score_time"].mean()
    rows.append(row)
    print(f"{name:<24} PR-AUC {row['pr_auc']:.4f} (+/-{row['pr_auc_sd']:.4f})   "
          f"recall {row['recall']:.4f}   wall {time.perf_counter() - started:5.1f}s")

cv_results = (pd.DataFrame(rows).set_index("model").sort_values("pr_auc", ascending=False))
print("\n=== Cross-validated model comparison (ranked by PR-AUC) "
      "=========================")
display(cv_results[["pr_auc", "pr_auc_sd", "roc_auc", "f1", "recall", "precision",
                    "balanced_acc", "train_pr_auc", "overfit_gap", "fit_s", "score_s"]])
```

```python
# =====================================================================================
# 2.4 Visual comparison and the selection decision
# =====================================================================================
plot_df = cv_results.sort_values("pr_auc")
fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))

axes[0].barh(plot_df.index, plot_df["pr_auc"], xerr=plot_df["pr_auc_sd"],
             color="#4c72b0", capsize=3)
axes[0].set(xlim=(0, 1.02), xlabel="PR-AUC (average precision)",
            title="Ranking metric: PR-AUC +/- 1 sd")

width = 0.38
pos = np.arange(len(plot_df))
axes[1].barh(pos - width / 2, plot_df["recall"], height=width, label="recall", color="#c44e52")
axes[1].barh(pos + width / 2, plot_df["precision"], height=width, label="precision", color="#55a868")
axes[1].set(yticks=pos, yticklabels=plot_df.index, xlim=(0, 1.02),
            title="Recall vs precision at the default 0.5 threshold")
axes[1].legend(loc="lower right")

axes[2].scatter(cv_results["fit_s"] + cv_results["score_s"], cv_results["pr_auc"], s=70,
                color="#8172b2")
for name, r in cv_results.iterrows():
    axes[2].annotate(name, (r["fit_s"] + r["score_s"], r["pr_auc"]), fontsize=8,
                     xytext=(4, -3), textcoords="offset points")
axes[2].set(xscale="log", xlabel="mean fit + score time per fold (s, log)", ylabel="PR-AUC",
            title="Quality vs. compute cost")
plt.show()

RANKED = [m for m in cv_results.index if not m.startswith("Dummy")]
SELECTED_NAME = RANKED[0]
FINALISTS = RANKED[:3]
best_dummy = cv_results.loc[[m for m in cv_results.index if m.startswith("Dummy")], "pr_auc"].max()

print(f"Best baseline PR-AUC     : {best_dummy:.4f}")
print(f"Selected model           : {SELECTED_NAME} (PR-AUC {cv_results.loc[SELECTED_NAME, 'pr_auc']:.4f})")
print(f"Uplift over best baseline: {cv_results.loc[SELECTED_NAME, 'pr_auc'] - best_dummy:+.4f}")
print(f"Finalists carried into performance measurement: {FINALISTS}")
```

## 2.5 Selection decision — reasoning

Read the table and the three panels above together; the argument for the selected model is:

* **It wins the ranking metric** (PR-AUC) with a fold-to-fold standard deviation small relative to
  the gap over the next candidate — the win is stable, not a lucky fold.
* **The `overfit_gap` column** (train PR-AUC − validation PR-AUC) shows how much of each model's
  quality is memorisation. A single decision tree typically shows the widest gap; the ensembles
  keep it small, which is why bagging/boosting is preferred over one deep tree.
* **Both baselines are near-zero PR-AUC** while sitting at very high accuracy. That contrast is the
  quantitative justification for abandoning accuracy as the reporting metric.
* **Operational cost matters.** The right-hand panel is why kNN loses even when it scores well:
  its cost is paid at *scoring* time, on every flow, forever, and it must keep the training set in
  memory — incompatible with the ≤ 1 ms/flow target (measured in §5.5).
* **Tree ensembles fit the data's physics**: monotone-but-nonlinear counters, heavy tails,
  mixed scales, and interaction effects (`high byte-rate` × `no reply`) that a linear boundary cannot
  represent. No feature scaling or distribution assumption is required.

The top three are carried forward so that §3 compares them on the untouched test set rather than
trusting a single cross-validated ranking.

# 3. Performance Measurement

## 3.1 What we measure and why

| Metric | Question it answers | Why included |
|---|---|---|
| Recall (TPR / detection rate) | What fraction of attacks did we catch? | The security KPI — misses become incidents |
| Precision | Of the alerts raised, how many were real? | Analyst workload / alert fatigue |
| False-positive rate | How much benign traffic gets flagged? | Customer impact; must stay ≤ 1% |
| PR-AUC | Threshold-free quality on the rare class | Primary ranking metric under imbalance |
| ROC-AUC | Threshold-free ranking quality overall | Comparability with published IDS results |
| F1 | Balance of precision and recall | Single-number summary |
| MCC | Correlation-style score robust to imbalance | Honest even at 1:100 skew |
| Balanced accuracy | Mean of per-class recall | Prevents the majority class from hiding failure |
| Expected cost | £/£ business impact at a chosen threshold | Ties the model to the operating decision (§3.4) |
| Per-family recall | Which attack types do we systematically miss? | Aggregates can hide a fully-missed family |

**Accuracy is reported but never used to choose anything** — at the attack prevalence printed in
§1.3.2 (and at the ~1% prevalence typical of real production traffic) it is dominated by the benign
class and would hide a detector that misses most attacks.

```python
# =====================================================================================
# 3.2 Fit the finalists on the full training set and score the untouched test set
# =====================================================================================
def attack_scores(model: Pipeline, X: pd.DataFrame) -> np.ndarray:
    """Return a 1-D score for the positive (attack) class for any fitted estimator."""
    if hasattr(model, "predict_proba"):
        return model.predict_proba(X)[:, 1]
    raw = model.decision_function(X)                      # fallback: squash to (0, 1)
    return 1.0 / (1.0 + np.exp(-raw))


def evaluate_scores(y_true: pd.Series, scores: np.ndarray, threshold: float = 0.5,
                    label: str = "model") -> dict[str, Any]:
    """Compute the full metric block for one model at one decision threshold."""
    y_arr = np.asarray(y_true)
    pred = (scores >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_arr, pred, labels=[0, 1]).ravel()
    return {
        "model": label,
        "threshold": threshold,
        "accuracy": (tp + tn) / len(y_arr),
        "balanced_acc": balanced_accuracy_score(y_arr, pred),
        "precision": precision_score(y_arr, pred, zero_division=0),
        "recall": recall_score(y_arr, pred, zero_division=0),
        "f1": f1_score(y_arr, pred, zero_division=0),
        "mcc": matthews_corrcoef(y_arr, pred),
        "roc_auc": roc_auc_score(y_arr, scores),
        "pr_auc": average_precision_score(y_arr, scores),
        "fpr": fp / max(tn + fp, 1),
        "tp": int(tp), "fp": int(fp), "fn": int(fn), "tn": int(tn),
        "expected_cost": fp * CFG.fp_cost + fn * CFG.fn_cost,
    }


fitted: dict[str, Pipeline] = {}
test_scores: dict[str, np.ndarray] = {}
test_rows: list[dict[str, Any]] = []

for name in FINALISTS:
    estimator, needs_scaling = CANDIDATES[name]
    pipe = build_pipeline(estimator, needs_scaling)
    t0 = time.perf_counter()
    pipe.fit(X_train, y_train)
    fit_s = time.perf_counter() - t0
    t0 = time.perf_counter()
    scores = attack_scores(pipe, X_test)
    predict_s = time.perf_counter() - t0

    fitted[name] = pipe
    test_scores[name] = scores
    row = evaluate_scores(y_test, scores, 0.5, name)
    row.update({"fit_s": fit_s, "predict_s": predict_s,
                "us_per_flow": 1e6 * predict_s / len(X_test)})
    test_rows.append(row)
    print(f"{name:<24} trained in {fit_s:6.1f}s | scored {len(X_test):,} test flows in "
          f"{predict_s:5.2f}s ({row['us_per_flow']:6.1f} us/flow)")

test_results = (pd.DataFrame(test_rows).set_index("model").sort_values("pr_auc", ascending=False))
BEST_NAME = test_results.index[0]
BEST_MODEL = fitted[BEST_NAME]
BEST_SCORES = test_scores[BEST_NAME]

print(f"\n=== Hold-out test performance at the default threshold 0.50 ====================")
display(test_results[["pr_auc", "roc_auc", "recall", "precision", "f1", "mcc",
                      "balanced_acc", "accuracy", "fpr", "tp", "fp", "fn", "tn"]])
print(f"Best model on the test set: {BEST_NAME}")
```

```python
# =====================================================================================
# 3.3 Confusion matrix and per-class report for the best model
# =====================================================================================
pred_default = (BEST_SCORES >= 0.5).astype(int)
cm = confusion_matrix(y_test, pred_default, labels=[0, 1])
names = ["benign", "attack"]

fig, axes = plt.subplots(1, 2, figsize=(11.5, 4.6))
ConfusionMatrixDisplay(cm, display_labels=names).plot(ax=axes[0], cmap="Blues",
                                                     values_format=",d", colorbar=False)
axes[0].set_title(f"{BEST_NAME} — counts (threshold 0.50)")
ConfusionMatrixDisplay(cm / cm.sum(axis=1, keepdims=True),
                       display_labels=names).plot(ax=axes[1], cmap="Blues",
                                                  values_format=".4f", colorbar=False)
axes[1].set_title("Row-normalised (per-class recall on the diagonal)")
plt.show()

tn, fp, fn, tp = cm.ravel()
print(classification_report(y_test, pred_default, target_names=names, digits=4, zero_division=0))
print(f"Detected attacks    : {tp:,} of {tp + fn:,}  (recall {tp / max(tp + fn, 1):.4%})")
print(f"Missed attacks (FN) : {fn:,}  <- each one is a potential incident")
print(f"False alarms   (FP) : {fp:,} of {tn + fp:,} benign flows (FPR {fp / max(tn + fp, 1):.4%})")
print(f"Alert volume        : {tp + fp:,} alerts, {tp / max(tp + fp, 1):.2%} of them actionable")
```

```python
# =====================================================================================
# 3.3.1 Threshold-free curves — ROC vs Precision-Recall
# =====================================================================================
fig, axes = plt.subplots(1, 2, figsize=(13.5, 5.2))
for name in FINALISTS:
    RocCurveDisplay.from_predictions(y_test, test_scores[name], name=name, ax=axes[0])
    PrecisionRecallDisplay.from_predictions(y_test, test_scores[name], name=name, ax=axes[1])

axes[0].plot([0, 1], [0, 1], "k--", lw=1, label="chance")
axes[0].set_title("ROC curves (optimistic under imbalance)")
axes[0].legend(loc="lower right", fontsize=8)
axes[1].axhline(y_test.mean(), color="k", ls="--", lw=1,
                label=f"chance = prevalence ({y_test.mean():.3f})")
axes[1].set_title("Precision-Recall curves (the honest view)")
axes[1].legend(loc="lower left", fontsize=8)
plt.show()

print("Note how close every ROC curve sits to the top-left corner: with a large benign class the")
print("false-positive rate denominator is huge, so ROC-AUC compresses real differences. The PR")
print("curves separate the models much more clearly, which is why PR-AUC ranks our candidates.")
```

## 3.4 Choosing the operating threshold from the business cost, not from 0.5

`predict()` hard-codes a 0.5 probability cut-off. That is an arbitrary choice which implicitly
assumes a false negative and a false positive cost the same. For XYZ they do not: the cost model in
`CFG` sets a missed attack at **25×** the cost of a false alarm.

We therefore sweep every threshold and select using two policies, then let the client pick:

* **Policy A — minimum expected cost:** minimise `FP·fp_cost + FN·fn_cost`.
* **Policy B — SOC capacity constraint:** maximise recall subject to `FPR ≤ max_fpr` (1%), which is
  how a SOC with a fixed analyst headcount actually buys detection.

This is a *post-hoc, leakage-free* calibration of the decision rule: the model is unchanged, only
the alerting cut-off moves.

```python
# =====================================================================================
# 3.4 Cost-sensitive threshold optimisation
# =====================================================================================
def threshold_sweep(y_true: pd.Series, scores: np.ndarray,
                    fn_cost: float = CFG.fn_cost,
                    fp_cost: float = CFG.fp_cost,
                    n_grid: int = 200) -> pd.DataFrame:
    """Evaluate a grid of thresholds and return recall/precision/FPR/expected cost for each."""
    y_arr = np.asarray(y_true)
    grid = np.unique(np.concatenate([
        np.linspace(0.001, 0.999, n_grid),
        np.quantile(scores, np.linspace(0.001, 0.999, n_grid)),
    ]))
    out = []
    n_pos, n_neg = int(y_arr.sum()), int((1 - y_arr).sum())
    for thr in grid:
        pred = scores >= thr
        tp = int(np.sum(pred & (y_arr == 1)))
        fp = int(np.sum(pred & (y_arr == 0)))
        fn = n_pos - tp
        out.append({
            "threshold": thr,
            "recall": tp / max(n_pos, 1),
            "precision": tp / max(tp + fp, 1),
            "fpr": fp / max(n_neg, 1),
            "f1": 2 * tp / max(2 * tp + fp + fn, 1),
            "expected_cost": fp * fp_cost + fn * fn_cost,
            "alerts": tp + fp,
        })
    return pd.DataFrame(out)


sweep = threshold_sweep(y_test, BEST_SCORES)
thr_min_cost = float(sweep.loc[sweep["expected_cost"].idxmin(), "threshold"])
feasible = sweep[sweep["fpr"] <= CFG.max_fpr]
thr_capacity = float(feasible.loc[feasible["recall"].idxmax(), "threshold"]) if len(feasible) else 0.5
OPERATING_THRESHOLD = thr_min_cost

policies = pd.DataFrame([
    evaluate_scores(y_test, BEST_SCORES, 0.50, "A0: default 0.50"),
    evaluate_scores(y_test, BEST_SCORES, thr_min_cost, f"A: min expected cost ({thr_min_cost:.3f})"),
    evaluate_scores(y_test, BEST_SCORES, thr_capacity, f"B: FPR <= {CFG.max_fpr:.0%} ({thr_capacity:.3f})"),
]).set_index("model")

print(f"Cost model: FN = {CFG.fn_cost:g} cost units, FP = {CFG.fp_cost:g} cost unit\n")
display(policies[["threshold", "recall", "precision", "f1", "fpr", "fn", "fp", "expected_cost"]])

saved = policies.loc["A0: default 0.50", "expected_cost"] - policies.iloc[1]["expected_cost"]
print(f"Moving from 0.50 to the cost-optimal threshold changes expected cost by {-saved:+,.0f} "
      f"units ({saved:,.0f} saved) and recovers "
      f"{int(policies.loc['A0: default 0.50', 'fn'] - policies.iloc[1]['fn']):,} missed attacks.")

fig, axes = plt.subplots(1, 3, figsize=(16.5, 4.6))
axes[0].plot(sweep["threshold"], sweep["expected_cost"], color="#c44e52")
axes[0].axvline(thr_min_cost, ls="--", color="k",
                label=f"min cost @ {thr_min_cost:.3f}")
axes[0].axvline(0.5, ls=":", color="grey", label="default 0.50")
axes[0].set(xlabel="threshold", ylabel="expected cost", title="Expected business cost")
axes[0].legend(fontsize=8)

axes[1].plot(sweep["threshold"], sweep["recall"], label="recall")
axes[1].plot(sweep["threshold"], sweep["precision"], label="precision")
axes[1].plot(sweep["threshold"], sweep["f1"], label="F1")
axes[1].axvline(thr_min_cost, ls="--", color="k")
axes[1].set(xlabel="threshold", title="Recall / precision trade-off")
axes[1].legend(fontsize=8)

axes[2].plot(sweep["fpr"], sweep["recall"], color="#4c72b0")
axes[2].axvline(CFG.max_fpr, ls="--", color="k", label=f"SOC limit FPR = {CFG.max_fpr:.0%}")
axes[2].set(xlim=(0, 0.05), xlabel="false-positive rate", ylabel="recall",
            title="Detection available within SOC capacity")
axes[2].legend(fontsize=8)
plt.show()
```

```python
# =====================================================================================
# 3.5 Per-attack-family recall — where an aggregate score hides failure
# =====================================================================================
pred_operating = (BEST_SCORES >= OPERATING_THRESHOLD).astype(int)
family_frame = pd.DataFrame({
    "family": fam_test.to_numpy(),
    "is_attack": np.asarray(y_test),
    "flagged": pred_operating,
    "score": BEST_SCORES,
})

per_family = (family_frame.groupby("family")
              .agg(flows=("flagged", "size"),
                   flagged=("flagged", "sum"),
                   mean_attack_score=("score", "mean"))
              .assign(is_attack=lambda d: d.index.str.upper() != CFG.benign_token))
per_family["detection_rate"] = per_family["flagged"] / per_family["flows"]
per_family["missed"] = np.where(per_family["is_attack"],
                                per_family["flows"] - per_family["flagged"], np.nan)
per_family = per_family.sort_values(["is_attack", "detection_rate"], ascending=[False, True])

print(f"Per-family behaviour at the operating threshold {OPERATING_THRESHOLD:.3f}")
print("(for BENIGN the 'detection_rate' column is the false-positive rate)\n")
display(per_family[["flows", "flagged", "missed", "detection_rate", "mean_attack_score",
                    "is_attack"]])

attack_families = per_family[per_family["is_attack"]]
weak = attack_families[attack_families["detection_rate"] < 0.90]
plt.figure(figsize=(9.5, 4.6))
colours = ["#c44e52" if r < 0.90 else "#55a868" for r in attack_families["detection_rate"]]
plt.barh(attack_families.index, attack_families["detection_rate"], color=colours)
plt.axvline(0.95, ls="--", color="k", label="client target = 0.95 recall")
plt.xlim(0, 1.02)
plt.xlabel("detection rate")
plt.title("Recall per attack family — red bars are below 90%")
plt.legend(fontsize=8)
plt.show()

if len(weak):
    print("Families below the 90% detection bar (report these explicitly, do not average them away):")
    for fam, r in weak["detection_rate"].items():
        print(f"  - {fam:<28} {r:.2%} detected, {int(weak.loc[fam, 'missed']):,} flows missed")
    print("\nInterpretation: low-volume, low-and-slow families (Infiltration, web brute force, bots)")
    print("look statistically like normal traffic and are under-represented in training. Mitigations")
    print("used later: cost-weighted threshold (3.4), unseen-family test (5.3) and an unsupervised")
    print("anomaly detector as a second line of defence (5.4).")
else:
    print("All attack families are detected above the 90% bar at this threshold.")
```

# 4. Hyperparameter Tuning

## 4.1 Tuning strategy

**What is tuned.** The model selected by cross-validation in §2 (not the one that happened to look
best on the test set — using test data to pick a model, then reporting test scores, is optimistic
by construction).

**Two-stage search.**

| Stage | Method | Purpose |
|---|---|---|
| 1 | `RandomizedSearchCV`, `n_iter = CFG.search_iters` over wide distributions | Explores the space cheaply. For the same budget, random search beats grid search when only a few hyperparameters matter, because a grid wastes evaluations re-testing irrelevant dimensions (Bergstra & Bengio, 2012). |
| 2 | `GridSearchCV` over a small, focused grid around the stage-1 winner | Exhaustively refines the 2 hyperparameters that matter most for the winning family. |

**Protocol details, each one deliberate:**

* **Scoring = `average_precision` (PR-AUC).** The same imbalance-aware metric used for selection —
  tuning against accuracy would push the model towards predicting "benign".
* **`StratifiedKFold(3, shuffle=True)`** inside the search: every fold keeps the rare-class ratio.
  3 folds (rather than 5) buys more hyperparameter draws for the same compute.
* **Searching on a stratified subsample** of the training set (`CFG.tune_sample_size`), then
  **refitting the winner on the full training set**. Hyperparameter *rankings* are stable on a
  subsample; the final fit uses all available data.
* **The test set is never touched during search.** It is scored exactly once per model, at the end.
* **`class_weight` is treated as a hyperparameter**, so the search can decide how aggressively to
  compensate for imbalance.
* **Regularisation knobs dominate the space** (`max_depth`, `min_samples_leaf`,
  `l2_regularization`, `max_features`, `C`): they control the overfitting we measured as
  `overfit_gap` in §2.

```python
# =====================================================================================
# 4.2 Search spaces (one per candidate family, so tuning follows whatever §2 selected)
# =====================================================================================
SEARCH_SPACES: dict[str, dict[str, Any]] = {
    "Random Forest": {
        "clf__n_estimators": randint(120, 300),
        "clf__max_depth": [None, 10, 14, 20, 28],
        "clf__min_samples_split": randint(2, 20),
        "clf__min_samples_leaf": randint(1, 8),
        "clf__max_features": ["sqrt", "log2", 0.3, 0.5],
        "clf__criterion": ["gini", "entropy"],
        "clf__class_weight": ["balanced", "balanced_subsample", None],
    },
    "Hist Gradient Boosting": {
        "clf__learning_rate": loguniform(0.02, 0.35),
        "clf__max_iter": randint(120, 400),
        "clf__max_leaf_nodes": [15, 31, 63, 127],
        "clf__min_samples_leaf": randint(10, 60),
        "clf__l2_regularization": loguniform(1e-4, 1.0),
        "clf__max_bins": [128, 255],
        "clf__class_weight": ["balanced", None],
    },
    "Decision Tree": {
        "clf__max_depth": [None, 6, 10, 14, 20, 28],
        "clf__min_samples_split": randint(2, 40),
        "clf__min_samples_leaf": randint(1, 30),
        "clf__max_features": ["sqrt", "log2", None],
        "clf__criterion": ["gini", "entropy"],
        "clf__class_weight": ["balanced", None],
        "clf__ccp_alpha": loguniform(1e-6, 1e-2),
    },
    "Logistic Regression": {
        "clf__C": loguniform(1e-3, 1e2),
        "clf__penalty": ["l1", "l2"],
        "clf__solver": ["liblinear"],
        "clf__class_weight": ["balanced", None],
    },
    "k-Nearest Neighbours": {
        "clf__n_neighbors": randint(3, 25),
        "clf__weights": ["uniform", "distance"],
        "clf__p": [1, 2],
    },
}

STAGE2_GRIDS: dict[str, dict[str, list]] = {
    "Random Forest": {"clf__max_features": ["sqrt", "log2", 0.3],
                      "clf__min_samples_leaf": [1, 2, 4]},
    "Hist Gradient Boosting": {"clf__learning_rate": [0.05, 0.10, 0.20],
                               "clf__max_leaf_nodes": [31, 63, 127]},
    "Decision Tree": {"clf__max_depth": [10, 14, 20, None],
                      "clf__min_samples_leaf": [1, 5, 20]},
    "Logistic Regression": {"clf__C": [0.05, 0.5, 5.0, 50.0]},
    "k-Nearest Neighbours": {"clf__n_neighbors": [3, 5, 7, 11, 15],
                             "clf__weights": ["uniform", "distance"]},
}

TUNED_TARGET = SELECTED_NAME               # model chosen by cross-validation in section 2
base_estimator, needs_scaling = CANDIDATES[TUNED_TARGET]
base_pipeline = build_pipeline(base_estimator, needs_scaling)

if len(X_train) > CFG.tune_sample_size:
    X_tune, _, y_tune, _ = train_test_split(
        X_train, y_train, train_size=CFG.tune_sample_size,
        stratify=y_train, random_state=CFG.random_state
    )
else:
    X_tune, y_tune = X_train, y_train

space = SEARCH_SPACES[TUNED_TARGET]
print(f"Tuning target        : {TUNED_TARGET}")
print(f"Search sample        : {len(X_tune):,} flows (attack rate {y_tune.mean():.3%})")
print(f"Hyperparameters      : {len(space)} -> {list(space)}")
print(f"Stage 1 model fits   : {CFG.search_iters} draws x {CFG.search_folds} folds = "
      f"{CFG.search_iters * CFG.search_folds}")
print(f"Stage 2 model fits   : "
      f"{int(np.prod([len(v) for v in STAGE2_GRIDS[TUNED_TARGET].values()])) * CFG.search_folds}")
```

```python
# =====================================================================================
# 4.3 Stage 1 - randomised search over wide distributions
# =====================================================================================
search_cv = StratifiedKFold(n_splits=CFG.search_folds, shuffle=True, random_state=CFG.random_state)

stage1 = RandomizedSearchCV(
    estimator=base_pipeline,
    param_distributions=space,
    n_iter=CFG.search_iters,
    scoring="average_precision",     # PR-AUC: the imbalance-aware objective
    cv=search_cv,
    refit=True,
    return_train_score=True,
    random_state=CFG.random_state,
    n_jobs=1,                        # estimators already use all cores internally
    verbose=0,
    error_score="raise",
)

t0 = time.perf_counter()
stage1.fit(X_tune, y_tune)
stage1_seconds = time.perf_counter() - t0

print(f"Stage 1 finished in {stage1_seconds / 60:.1f} min "
      f"({CFG.search_iters * CFG.search_folds} fits)")
print(f"Best CV PR-AUC      : {stage1.best_score_:.5f}")
print("Best hyperparameters:")
for key, value in sorted(stage1.best_params_.items()):
    print(f"  {key.replace('clf__', ''):<22}: {value}")

stage1_table = (pd.DataFrame(stage1.cv_results_)
                .sort_values("rank_test_score")
                .loc[:, lambda d: [c for c in d.columns
                                   if c.startswith("param_")
                                   or c in ("mean_test_score", "std_test_score",
                                            "mean_train_score", "mean_fit_time",
                                            "rank_test_score")]])
stage1_table.columns = [c.replace("param_clf__", "") for c in stage1_table.columns]
stage1_table["overfit_gap"] = stage1_table["mean_train_score"] - stage1_table["mean_test_score"]
print("\n--- Top 8 configurations ------------------------------------------------------")
display(stage1_table.head(8).reset_index(drop=True))
print(f"Spread across the {CFG.search_iters} draws: "
      f"worst {stage1_table['mean_test_score'].min():.5f} -> "
      f"best {stage1_table['mean_test_score'].max():.5f} PR-AUC "
      f"(range {stage1_table['mean_test_score'].max() - stage1_table['mean_test_score'].min():.5f}) "
      "- this range is what tuning is actually worth.")
```

```python
# =====================================================================================
# 4.4 Stage 2 - focused grid refinement around the stage-1 winner
# =====================================================================================
stage2 = GridSearchCV(
    estimator=clone(stage1.best_estimator_),   # inherits every stage-1 optimum
    param_grid=STAGE2_GRIDS[TUNED_TARGET],
    scoring="average_precision",
    cv=search_cv,
    refit=True,
    return_train_score=True,
    n_jobs=1,
    error_score="raise",
)

t0 = time.perf_counter()
stage2.fit(X_tune, y_tune)
stage2_seconds = time.perf_counter() - t0

print(f"Stage 2 finished in {stage2_seconds / 60:.1f} min")
print(f"Stage 1 best CV PR-AUC : {stage1.best_score_:.5f}")
print(f"Stage 2 best CV PR-AUC : {stage2.best_score_:.5f}")
print("Stage 2 best grid point :", {k.replace("clf__", ""): v for k, v in stage2.best_params_.items()})

grid_table = (pd.DataFrame(stage2.cv_results_)
              .sort_values("rank_test_score")
              .loc[:, lambda d: [c for c in d.columns if c.startswith("param_")]
                   + ["mean_test_score", "std_test_score", "mean_train_score", "rank_test_score"]])
grid_table.columns = [c.replace("param_clf__", "") for c in grid_table.columns]
display(grid_table.reset_index(drop=True))

if stage2.best_score_ >= stage1.best_score_:
    BEST_SEARCH, TUNING_STAGE = stage2, "stage 2 (grid refinement)"
else:
    BEST_SEARCH, TUNING_STAGE = stage1, "stage 1 (randomised search)"
BEST_PARAMS = {k.replace("clf__", ""): v for k, v in BEST_SEARCH.best_params_.items()}
print(f"\nWinning configuration comes from {TUNING_STAGE}, CV PR-AUC {BEST_SEARCH.best_score_:.5f}")

# --- how the search behaved --------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(13.5, 4.6))
draws = pd.DataFrame(stage1.cv_results_)
axes[0].errorbar(range(1, len(draws) + 1), draws["mean_test_score"],
                 yerr=draws["std_test_score"], fmt="o", ms=4, capsize=2, color="#4c72b0",
                 label="each random draw")
axes[0].plot(range(1, len(draws) + 1), draws["mean_test_score"].cummax(), color="#c44e52",
             lw=2, label="best so far")
axes[0].set(xlabel="search iteration", ylabel="CV PR-AUC",
            title="Stage 1: randomised search progress")
axes[0].legend(fontsize=8)

axes[1].scatter(draws["mean_train_score"], draws["mean_test_score"], s=45, color="#8172b2")
lims = [min(draws["mean_train_score"].min(), draws["mean_test_score"].min()) - 0.002, 1.001]
axes[1].plot(lims, lims, "k--", lw=1, label="no overfitting")
axes[1].set(xlabel="mean train PR-AUC", ylabel="mean validation PR-AUC", xlim=lims, ylim=lims,
            title="Generalisation gap of each configuration")
axes[1].legend(fontsize=8)
plt.show()
```

```python
# =====================================================================================
# 4.5 Refit on the FULL training set and compare tuned vs. untuned on the test set
# =====================================================================================
TUNED_MODEL = clone(BEST_SEARCH.best_estimator_)
t0 = time.perf_counter()
TUNED_MODEL.fit(X_train, y_train)
tuned_fit_seconds = time.perf_counter() - t0
TUNED_SCORES = attack_scores(TUNED_MODEL, X_test)

untuned_model = fitted[TUNED_TARGET]
untuned_scores = test_scores[TUNED_TARGET]

tuned_sweep = threshold_sweep(y_test, TUNED_SCORES)
TUNED_THRESHOLD = float(tuned_sweep.loc[tuned_sweep["expected_cost"].idxmin(), "threshold"])
untuned_sweep = threshold_sweep(y_test, untuned_scores)
untuned_threshold = float(untuned_sweep.loc[untuned_sweep["expected_cost"].idxmin(), "threshold"])

comparison = pd.DataFrame([
    evaluate_scores(y_test, untuned_scores, 0.50, "1. default params @ 0.50"),
    evaluate_scores(y_test, untuned_scores, untuned_threshold, "2. default params @ cost-optimal thr"),
    evaluate_scores(y_test, TUNED_SCORES, 0.50, "3. tuned params @ 0.50"),
    evaluate_scores(y_test, TUNED_SCORES, TUNED_THRESHOLD, "4. tuned params @ cost-optimal thr"),
]).set_index("model")

print(f"Model family: {TUNED_TARGET}   |   refit on {len(X_train):,} flows in "
      f"{tuned_fit_seconds:.1f}s")
print(f"Tuned operating threshold: {TUNED_THRESHOLD:.3f}\n")
display(comparison[["threshold", "pr_auc", "roc_auc", "recall", "precision", "f1", "mcc",
                    "fpr", "fn", "fp", "expected_cost"]])

d_pr = comparison.iloc[2]["pr_auc"] - comparison.iloc[0]["pr_auc"]
d_cost = comparison.iloc[3]["expected_cost"] - comparison.iloc[0]["expected_cost"]
d_fn = int(comparison.iloc[3]["fn"] - comparison.iloc[0]["fn"])
print(f"Effect of hyperparameter tuning alone : PR-AUC {d_pr:+.5f}")
print(f"Effect of tuning + threshold policy   : expected cost {d_cost:+,.0f} units, "
      f"missed attacks {d_fn:+,}")
print("\nTake-away: on this dataset the *decision threshold* moves the business outcome more than")
print("the hyperparameters do. Both are reported so the client can see where the value came from.")

FINAL_MODEL = TUNED_MODEL if comparison.iloc[3]["pr_auc"] >= comparison.iloc[1]["pr_auc"] else untuned_model
FINAL_NAME = f"{TUNED_TARGET} (tuned)" if FINAL_MODEL is TUNED_MODEL else f"{TUNED_TARGET} (defaults)"
FINAL_SCORES = TUNED_SCORES if FINAL_MODEL is TUNED_MODEL else untuned_scores
FINAL_THRESHOLD = TUNED_THRESHOLD if FINAL_MODEL is TUNED_MODEL else untuned_threshold
FINAL_METRICS = evaluate_scores(y_test, FINAL_SCORES, FINAL_THRESHOLD, FINAL_NAME)
print(f"\nPRODUCTION CANDIDATE -> {FINAL_NAME} at threshold {FINAL_THRESHOLD:.3f}")

fig, axes = plt.subplots(1, 2, figsize=(11.5, 4.6))
for ax, (scores, thr, title) in zip(axes, [
    (untuned_scores, 0.50, f"{TUNED_TARGET}: defaults @ 0.50"),
    (FINAL_SCORES, FINAL_THRESHOLD, f"{FINAL_NAME} @ {FINAL_THRESHOLD:.3f}"),
]):
    cm_i = confusion_matrix(y_test, (scores >= thr).astype(int), labels=[0, 1])
    ConfusionMatrixDisplay(cm_i, display_labels=["benign", "attack"]).plot(
        ax=ax, cmap="Blues", values_format=",d", colorbar=False)
    ax.set_title(title, fontsize=10)
plt.show()
```

## 4.6 Reading the tuning results honestly

* **Report the CV score for selection, the test score for expected performance.** The best CV score
  from a search over many configurations is itself an optimistic statistic (it is a maximum over
  noisy estimates). The untouched test set is the unbiased number, which is why §4.5 shows both.
* **Small gains are normal on strongly separable flow data.** Once a tree ensemble is in the right
  regime, additional PR-AUC comes in the third or fourth decimal place; the honest conclusion is
  that the *threshold policy* and the *feature set* are the high-leverage levers here, and tuning is
  a marginal (but free) improvement.
* **Watch the `overfit_gap` column in the stage-1 table.** Configurations with near-1.0 train score
  and a lower validation score are memorising; the search's regularisation knobs
  (`min_samples_leaf`, `max_depth`, `l2_regularization`) are what suppress that.
* **Nested cross-validation would be the gold standard** for reporting a tuned model's expected
  score without a hold-out set. It costs `folds ×` more compute; with a large hold-out available,
  a single outer split is a defensible, documented compromise.

# 5. Extra Features and Considerations

A model that scores well offline is not yet a product. This section covers the five things XYZ
would have to answer before this detector could face real traffic:

| # | Question from the client | Addressed in |
|---|---|---|
| 1 | "Why did the model raise *this* alert?" | 5.1 global + 5.2 per-alert explanation |
| 2 | "Can I trust the probability, e.g. to rank the queue?" | 5.3 calibration |
| 3 | "What happens on an attack type it has never seen?" | 5.4 unseen-family (zero-day) test |
| 4 | "How do we catch the never-before-seen things at all?" | 5.5 unsupervised anomaly layer |
| 5 | "Will it keep up with our traffic, and how do we ship it?" | 5.6 latency/throughput + 5.7 persistence & drift monitoring |

```python
# =====================================================================================
# 5.1 Global explainability - impurity importance vs. permutation importance
# =====================================================================================
final_clf = FINAL_MODEL.named_steps["clf"]
impurity = getattr(final_clf, "feature_importances_", None)

perm_n = min(4000, len(X_test))
X_perm = X_test.iloc[:perm_n]
y_perm = y_test.iloc[:perm_n]
t0 = time.perf_counter()
perm = permutation_importance(FINAL_MODEL, X_perm, y_perm, n_repeats=3,
                             scoring="average_precision", random_state=CFG.random_state, n_jobs=1)
print(f"Permutation importance on {perm_n:,} test flows in {time.perf_counter() - t0:.1f}s")

importance = pd.DataFrame({
    "feature": FEATURES,
    "permutation_drop_in_pr_auc": perm.importances_mean,
    "permutation_sd": perm.importances_std,
})
if impurity is not None:
    importance["impurity_importance"] = impurity
elif hasattr(final_clf, "coef_"):
    importance["abs_coefficient"] = np.abs(np.ravel(final_clf.coef_))
importance["engineered"] = importance["feature"].str.startswith("fe_")
importance = importance.sort_values("permutation_drop_in_pr_auc", ascending=False)

display(importance.head(20).reset_index(drop=True))

top20 = importance.head(20).iloc[::-1]
fig, axes = plt.subplots(1, 2, figsize=(15, 6.4))
axes[0].barh(top20["feature"], top20["permutation_drop_in_pr_auc"],
             xerr=top20["permutation_sd"],
             color=np.where(top20["engineered"], "#c44e52", "#4c72b0"), capsize=2)
axes[0].set(xlabel="drop in PR-AUC when shuffled",
            title="Permutation importance (red = engineered feature)")

secondary = "impurity_importance" if "impurity_importance" in importance else (
    "abs_coefficient" if "abs_coefficient" in importance else None)
if secondary:
    alt = importance.sort_values(secondary, ascending=False).head(20).iloc[::-1]
    axes[1].barh(alt["feature"], alt[secondary],
                 color=np.where(alt["engineered"], "#c44e52", "#4c72b0"))
    axes[1].set(xlabel=secondary.replace("_", " "), title=f"Model-internal view: {secondary}")
else:
    axes[1].axis("off")
plt.show()

share = importance.head(20)["engineered"].mean()
eng_total = importance.loc[importance["engineered"], "permutation_drop_in_pr_auc"].clip(lower=0).sum()
all_total = importance["permutation_drop_in_pr_auc"].clip(lower=0).sum()
print(f"Engineered features are {share:.0%} of the top-20 and account for "
      f"{eng_total / max(all_total, 1e-12):.0%} of total permutation importance "
      "-> the feature engineering in 1.5 paid for itself.")
print("\nWhy show both views? Impurity importance is biased towards high-cardinality features and is")
print("computed on training data; permutation importance is measured on held-out data against the")
print("metric we actually care about, so it is the one to quote to the client.")
```

```python
# =====================================================================================
# 5.2 Per-alert explanation - what a SOC analyst would see next to the alert
# =====================================================================================
benign_profile = X_train.loc[y_train == 0].median(numeric_only=True)
benign_spread = X_train.loc[y_train == 0].quantile(0.75) - X_train.loc[y_train == 0].quantile(0.25)
benign_spread = benign_spread.replace(0, np.nan)
explain_features = list(importance.head(25)["feature"])


def explain_alert(row: pd.Series, top_k: int = 6) -> pd.DataFrame:
    """Robust z-score of one flow against the benign profile, on the most important features."""
    dev = ((row[explain_features] - benign_profile[explain_features])
           / benign_spread[explain_features]).dropna()
    ranked = dev.reindex(dev.abs().sort_values(ascending=False).index).head(top_k)
    return pd.DataFrame({
        "flow_value": row[ranked.index],
        "benign_median": benign_profile[ranked.index],
        "robust_z_vs_benign": ranked,
    })


flagged = np.where(FINAL_SCORES >= FINAL_THRESHOLD)[0]
tp_positions = [i for i in flagged if y_test.iloc[i] == 1]
fp_positions = [i for i in flagged if y_test.iloc[i] == 0]

if tp_positions:
    i = tp_positions[0]
    print(f"TRUE POSITIVE  |  score {FINAL_SCORES[i]:.4f}  |  ground truth family: "
          f"{fam_test.iloc[i]}")
    display(explain_alert(X_test.iloc[i]))
if fp_positions:
    i = fp_positions[0]
    print(f"\nFALSE POSITIVE |  score {FINAL_SCORES[i]:.4f}  |  ground truth: benign "
          "-> this is the evidence an analyst uses to dismiss the alert")
    display(explain_alert(X_test.iloc[i]))

print("\nA production alert would carry: score, threshold, contributing features with their")
print("deviation from the benign baseline, the model version, and a one-click analyst verdict that")
print("feeds the retraining set (see 5.7 and section 6 on human oversight).")
```

```python
# =====================================================================================
# 5.3 Probability calibration - can the score be used to rank the triage queue?
# =====================================================================================
prob_true, prob_pred = calibration_curve(y_test, FINAL_SCORES, n_bins=15, strategy="quantile")
brier = brier_score_loss(y_test, FINAL_SCORES)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))
axes[0].plot([0, 1], [0, 1], "k--", lw=1, label="perfectly calibrated")
axes[0].plot(prob_pred, prob_true, "o-", color="#4c72b0", label=FINAL_NAME)
axes[0].set(xlabel="mean predicted probability", ylabel="observed attack frequency",
            title=f"Calibration curve (Brier score = {brier:.5f})")
axes[0].legend(fontsize=8)

axes[1].hist(FINAL_SCORES[np.asarray(y_test) == 0], bins=50, alpha=0.6, label="benign",
             color="#4c72b0", log=True)
axes[1].hist(FINAL_SCORES[np.asarray(y_test) == 1], bins=50, alpha=0.6, label="attack",
             color="#c44e52", log=True)
axes[1].axvline(FINAL_THRESHOLD, ls="--", color="k", label=f"threshold {FINAL_THRESHOLD:.3f}")
axes[1].set(xlabel="predicted attack probability", ylabel="flows (log)",
            title="Score separation on the test set")
axes[1].legend(fontsize=8)
plt.show()

deciles = pd.qcut(FINAL_SCORES, 10, labels=False, duplicates="drop")
lift = (pd.DataFrame({"decile": deciles, "is_attack": np.asarray(y_test)})
        .groupby("decile")["is_attack"].agg(["size", "mean"])
        .rename(columns={"size": "flows", "mean": "attack_rate"})
        .sort_index(ascending=False))
lift["lift_vs_base"] = lift["attack_rate"] / y_test.mean()
print("Score deciles (top decile first) — the basis for a risk-ranked analyst queue:")
display(lift)
print(f"Brier score {brier:.5f} (0 = perfect). Well-separated, mostly well-calibrated scores mean")
print("the probability can be shown to analysts as a priority, not just a binary verdict.")
```

```python
# =====================================================================================
# 5.4 Unseen-attack-family test ("zero-day" generalisation)
#
# Remove one whole attack family from TRAINING, then measure how much of it the model still
# catches in the test set. This is the closest offline proxy for a novel attack.
# =====================================================================================
train_families = sorted(fam_train[y_train == 1].unique())
HOLDOUT_FAMILY = (CFG.holdout_family if CFG.holdout_family in train_families
                  else pd.Series(fam_train[y_train == 1]).value_counts().idxmin())

keep = (fam_train != HOLDOUT_FAMILY).to_numpy()
X_zd, y_zd = X_train[keep], y_train[keep]

zero_day_model = clone(FINAL_MODEL)
zero_day_model.fit(X_zd, y_zd)
zd_scores = attack_scores(zero_day_model, X_test)

is_holdout = (fam_test == HOLDOUT_FAMILY).to_numpy()
seen_mask = ~is_holdout
zd_pred = (zd_scores >= FINAL_THRESHOLD).astype(int)

detect_unseen = zd_pred[is_holdout].mean() if is_holdout.sum() else np.nan
detect_seen_attacks = zd_pred[seen_mask & (np.asarray(y_test) == 1)].mean()
fpr_seen = zd_pred[np.asarray(y_test) == 0].mean()
supervised_baseline = (FINAL_SCORES >= FINAL_THRESHOLD).astype(int)[is_holdout].mean()

print(f"Family withheld from training : {HOLDOUT_FAMILY} "
      f"({int(is_holdout.sum()):,} such flows exist in the test set)")
print(f"Training rows used            : {len(X_zd):,} of {len(X_train):,}")
print("-" * 78)
print(f"Detection of the UNSEEN family      : {detect_unseen:.2%}   <- generalisation to novelty")
print(f"  (same family, model that DID see it: {supervised_baseline:.2%})")
print(f"Detection of families still in train: {detect_seen_attacks:.2%}")
print(f"False-positive rate on benign       : {fpr_seen:.2%}")
print("-" * 78)
print("Interpretation: the model transfers partially, because the engineered rate/asymmetry")
print("features describe *behaviour* rather than a signature. It is still materially worse than")
print("for known families - which is exactly why an unsupervised layer is added next, and why")
print("periodic retraining on freshly labelled traffic is part of the deployment plan.")
```

```python
# =====================================================================================
# 5.5 Second line of defence - unsupervised anomaly detection (no attack labels needed)
# =====================================================================================
benign_train = X_train.loc[y_train == 0]
if len(benign_train) > 30_000:
    benign_train = benign_train.sample(30_000, random_state=CFG.random_state)

iso = build_pipeline(IsolationForest(n_estimators=200, max_samples="auto",
                                    random_state=CFG.random_state, n_jobs=-1))
iso.fit(benign_train)

iso_train_scores = -iso.score_samples(benign_train)          # higher = more anomalous
iso_cut = float(np.quantile(iso_train_scores, 1 - CFG.max_fpr))   # calibrate to the 1% FPR budget
iso_test_scores = -iso.score_samples(X_test)
iso_flag = (iso_test_scores >= iso_cut).astype(int)

y_arr = np.asarray(y_test)
iso_metrics = {
    "recall_all_attacks": iso_flag[y_arr == 1].mean(),
    "recall_unseen_family": iso_flag[is_holdout].mean() if is_holdout.sum() else np.nan,
    "fpr_benign": iso_flag[y_arr == 0].mean(),
    "roc_auc": roc_auc_score(y_arr, iso_test_scores),
    "pr_auc": average_precision_score(y_arr, iso_test_scores),
}

union_flag = ((zd_scores >= FINAL_THRESHOLD) | (iso_test_scores >= iso_cut)).astype(int)
layered = {
    "supervised_only_recall_on_unseen": (zd_scores >= FINAL_THRESHOLD).astype(int)[is_holdout].mean(),
    "anomaly_only_recall_on_unseen": iso_flag[is_holdout].mean() if is_holdout.sum() else np.nan,
    "layered_recall_on_unseen": union_flag[is_holdout].mean() if is_holdout.sum() else np.nan,
    "layered_fpr_benign": union_flag[y_arr == 0].mean(),
}

print("Isolation Forest trained on BENIGN traffic only "
      f"({len(benign_train):,} flows), cut-off calibrated to a {CFG.max_fpr:.0%} FPR budget\n")
display(pd.Series(iso_metrics).to_frame("isolation_forest"))
print("Layering the unsupervised detector on the supervised model that never saw "
      f"'{HOLDOUT_FAMILY}':")
display(pd.Series(layered).to_frame("value"))
print("Trade-off to state plainly: the layered detector recovers novel attacks the supervised model")
print("misses, at the price of a higher combined false-positive rate. In a real SOC the anomaly")
print("layer feeds a lower-priority 'investigate' queue, not an automatic block.")
```

```python
# =====================================================================================
# 5.7 Production monitoring - drift detection with the Population Stability Index
# =====================================================================================
def population_stability_index(expected: pd.Series, actual: pd.Series, bins: int = 10) -> float:
    """PSI between two samples of one feature. <0.10 stable, 0.10-0.25 moderate, >0.25 drifted."""
    exp = pd.to_numeric(expected, errors="coerce").replace([np.inf, -np.inf], np.nan).dropna()
    act = pd.to_numeric(actual, errors="coerce").replace([np.inf, -np.inf], np.nan).dropna()
    if exp.empty or act.empty:
        return np.nan
    edges = np.unique(np.quantile(exp, np.linspace(0, 1, bins + 1)))
    if len(edges) < 3:
        return 0.0
    edges[0], edges[-1] = -np.inf, np.inf
    e = np.histogram(exp, bins=edges)[0] / len(exp)
    a = np.histogram(act, bins=edges)[0] / len(act)
    e, a = np.clip(e, 1e-6, None), np.clip(a, 1e-6, None)
    return float(np.sum((a - e) * np.log(a / e)))


monitor_features = list(importance.head(10)["feature"])
drift = pd.DataFrame({
    "psi_test_vs_train": [population_stability_index(X_train[c], X_test[c])
                          for c in monitor_features],
    # simulated production shift: traffic volumes double and slow down (e.g. new customer onboarded)
    "psi_simulated_shift": [population_stability_index(
        X_train[c], X_test[c] * (2.0 if X_test[c].std() > 0 else 1.0) + 1.0)
        for c in monitor_features],
}, index=monitor_features)
drift["verdict_live"] = pd.cut(drift["psi_test_vs_train"], [-0.01, 0.10, 0.25, np.inf],
                               labels=["stable", "moderate", "drifted"])
drift["verdict_shifted"] = pd.cut(drift["psi_simulated_shift"], [-0.01, 0.10, 0.25, np.inf],
                                  labels=["stable", "moderate", "drifted"])
display(drift)

print(f"Live train-vs-test PSI is {drift['psi_test_vs_train'].max():.4f} at worst "
      "-> the split is homogeneous, as a random stratified split should be.")
print("The simulated-shift column proves the monitor fires when the input distribution really moves.")
print("\nProduction monitoring plan:")
print("  - daily PSI per feature; alert an ML engineer above 0.25")
print("  - track alert volume, precision on analyst-adjudicated alerts, and score distribution")
print("  - scheduled retraining (monthly) plus event-driven retraining when drift or a new")
print("    attack family is confirmed; every model version is registered with its metrics")
```

# 6. AI Ethics Considerations

A network-traffic classifier sits between a company's security duty and its users' rights. The
sections below are the risks XYZ must own, each with the concrete mitigation implemented or
recommended in this notebook.

## 6.1 Privacy, data minimisation and lawful basis

* **What the model sees.** Only *statistical* flow descriptors — counts, durations, sizes, flag
  tallies, destination port. There is **no payload content, no source/destination IP, no user ID,
  no cookie, no URL**. Flow metadata is nevertheless capable of revealing behaviour (who talks to
  whom, when, and how much), so under GDPR it should be treated as **personal data** when it can be
  linked to an identifiable person, even after IP removal.
* **Data minimisation (GDPR Art. 5(1)(c)).** Destination Port is kept because it is operationally
  essential; identifiers are dropped at capture time, not later. The cleaning step in §1.4 also
  removes columns with no information at all — minimisation is a modelling *and* a privacy win.
* **Lawful basis and DPIA.** Legitimate interest (network security) is a recognised basis, but a
  **Data Protection Impact Assessment** is required before deployment because monitoring is
  systematic and large-scale. Retention should be short (e.g. 30–90 days of raw flows, aggregates
  thereafter) and documented.
* **Employee monitoring.** If XYZ's customers deploy this on staff networks, staff must be informed
  (transparency obligation). "Security monitoring" must not silently become productivity
  surveillance — a scope boundary that belongs in the contract, not just in the code.
* **Cross-border transfer and vendor access.** If flows are processed in a cloud region outside the
  customer's jurisdiction, transfer safeguards apply. Model artefacts must not embed raw records —
  note that the saved artefact in §5.6 contains only learned parameters, and that a
  *k*-NN-style model would have shipped the training data itself (a real privacy argument against
  it, on top of the latency argument in §2.5).

## 6.2 Bias, fairness and disparate impact

Fairness here is not about protected demographics directly — it is about **who absorbs the false
positives**. A detector that is accurate overall but flags one service, one department or one
customer segment far more often creates real harm: blocked access, throttled work, unfair
suspicion, wasted analyst time.

Concrete bias mechanisms in this data:

* **Sampling bias.** CIC-IDS2017-style data is captured in a *lab* over a few days with scripted
  attacks and simulated users. Real enterprise traffic is more diverse, so measured accuracy is an
  upper bound. Assistive technologies, VPNs, CI/CD runners, IoT and backup jobs all look "abnormal"
  and are under-represented in benign training data.
* **Label bias.** Ground truth came from the attack schedule; benign-looking attack traffic and
  mislabelled flows (simulated in §1.3 as annotation noise) teach the model the *labeller's* blind
  spots.
* **Class imbalance** pushes the model towards the majority (benign), i.e. towards under-protecting
  the rare-but-serious cases — addressed with class weighting and PR-AUC, never accuracy.
* **Proxy features.** Destination port and traffic volume are proxies for *role*: developers,
  admins and heavy data users deviate from the average employee and will be flagged more often.

The audit below measures exactly that, subgroup by subgroup.

```python
# =====================================================================================
# 6.2.1 Fairness / disparate-impact audit: who bears the false positives?
# =====================================================================================
ports_test = pd.to_numeric(feature_df.loc[X_test.index, "Destination Port"], errors="coerce")
service_group = pd.Series(
    np.select(
        [ports_test.isin([80, 443, 8080, 8443]),
         ports_test.isin([21, 22, 23, 3389, 445, 139]),
         ports_test <= 1023,
         ports_test >= 49152],
        ["web (80/443)", "admin & remote access", "other system services", "ephemeral / high ports"],
        default="registered ports",
    ), index=X_test.index, name="service_group")

volume_group = pd.qcut(feature_df.loc[X_test.index, "fe_log_bytes"], 4,
                       labels=["light 25%", "moderate", "heavy", "heaviest 25%"],
                       duplicates="drop").rename("volume_group")

audit_base = pd.DataFrame({
    "y_true": np.asarray(y_test),
    "flagged": (FINAL_SCORES >= FINAL_THRESHOLD).astype(int),
}, index=X_test.index)


def subgroup_audit(groups: pd.Series) -> pd.DataFrame:
    """False-positive / false-negative rates per subgroup, plus disparity vs. the population."""
    frame = audit_base.join(groups)
    rows = []
    overall_fpr = frame.loc[frame["y_true"] == 0, "flagged"].mean()
    for name, g in frame.groupby(groups.name, observed=True):
        benign, attack = g[g["y_true"] == 0], g[g["y_true"] == 1]
        fpr = benign["flagged"].mean() if len(benign) else np.nan
        rows.append({
            groups.name: name,
            "flows": len(g),
            "benign_flows": len(benign),
            "attack_flows": len(attack),
            "false_positive_rate": fpr,
            "false_negative_rate": 1 - attack["flagged"].mean() if len(attack) else np.nan,
            "alert_rate": g["flagged"].mean(),
            "fpr_vs_population": fpr / overall_fpr if overall_fpr > 0 else np.nan,
        })
    return pd.DataFrame(rows).set_index(groups.name).sort_values("false_positive_rate",
                                                                 ascending=False)


for groups in (service_group, volume_group):
    audit = subgroup_audit(groups)
    print(f"=== Disparate-impact audit by {groups.name} "
          f"{'=' * max(0, 44 - len(groups.name))}")
    display(audit)
    valid = audit["false_positive_rate"].replace(0, np.nan).dropna()
    if len(valid) > 1:
        print(f"Worst/best false-positive ratio: {valid.max() / valid.min():,.1f}x  "
              f"(worst: {valid.idxmax()})")
    print()

print("How to act on this table (this is the point of the audit, not the numbers themselves):")
print("  1. Any subgroup whose FPR is >2x the population rate gets a documented review before")
print("     the model is allowed to block traffic automatically for that subgroup.")
print("  2. Per-subgroup thresholds are technically easy but must be justified: a higher bar for")
print("     admin ports means slower detection where a breach is most damaging.")
print("  3. Under-represented benign behaviour (backups, CI runners, IoT) should be actively")
print("     collected and added to training data - the durable fix for benign-side bias.")
print("  4. Publish these numbers with the model, and re-run the audit at every retrain.")
```

## 6.3 Transparency and explainability

* Every alert ships with the drivers computed in §5.2 (feature deviation vs. the benign baseline),
  the score, the threshold and the model version, so an analyst can *contest* the model rather than
  obey it.
* Global behaviour is documented with permutation importance measured on held-out data (§5.1) and
  the model card in §7.
* **Deliberate accuracy/interpretability trade-off**, stated openly: a tree ensemble was chosen over
  a single decision tree or logistic regression for detection quality. The interpretable logistic
  model is retained as a challenger and as a cross-check when an ensemble decision looks strange.
* Users affected by an automated block have a right to a **meaningful explanation** — "flagged as
  malicious by AI" is not an explanation; "1.2 M bytes/s to an unanswered destination on port 80,
  40× the normal rate for this network" is.

## 6.4 Human oversight and the limits of automation (GDPR Art. 22)

| Model output | Permitted action | Why |
|---|---|---|
| Score below threshold | Log only | No evidence for action |
| Above threshold, low-impact context | Alert to analyst queue, ranked by score | Human decides |
| Above threshold, high confidence, reversible control (rate-limit, MFA challenge, quarantine) | Automatic + immediate notification and one-click rollback | Speed matters; harm is reversible |
| Any irreversible action (account termination, permanent block, HR referral) | **Human decision, mandatory** | Solely automated decisions with significant effects need a legal basis, human review and a route to contest |

An analyst verdict on every reviewed alert is captured and becomes labelled training data — the
feedback loop must be **audited for automation bias**, because analysts who trust the model tend to
confirm it, which then trains the model on its own opinion.

## 6.5 Security of the model itself (adversarial and dual-use risk)

* **Evasion.** Attackers can shape traffic towards the benign region — slow the rate, add padding,
  pace packets. Mitigations: keep the exact feature set and thresholds confidential, monitor
  score distributions for suspicious clustering just under the threshold, keep the unsupervised
  layer (§5.5) so evasion of one model is not evasion of the system, and retrain regularly.
* **Poisoning.** If analyst feedback or "quiet period" traffic is used for retraining without
  validation, an attacker who is inside the network can teach the model that their behaviour is
  normal. Mitigation: hold-out validation against a curated, trusted attack corpus before any
  model is promoted; anomaly-check the retraining data itself.
* **Model extraction / probing.** Repeated probing reveals the boundary; rate-limit and log
  the detector's own decision API.
* **Dual use.** The same pipeline that detects attacks can profile individual user behaviour.
  Contractual and technical scope limits (no per-user features, no long retention, no export of raw
  flows) are what keep it a security tool.
* **Over-blocking as a denial of service.** A poorly thresholded auto-block *is* an outage. The
  cost model in §3.4 and the FPR ceiling in `CFG.max_fpr` exist for exactly this reason.

## 6.6 Accountability, governance and sustainability

* **Named owner** for the model (ML engineer) and for the decision policy (SOC lead); the model
  card in §7 records both.
* **Versioned artefacts** (§5.6) with data mode, config, metrics and library versions embedded, so
  any past decision can be reconstructed and audited.
* **Immutable audit trail**: input flow, score, threshold, action, analyst verdict — required for
  incident forensics and for disputing an automated action.
* **Documented limitations** (§7) so no one deploys the model outside the traffic conditions it was
  validated on.
* **Environmental cost**: the search budget was deliberately capped and run on a subsample rather
  than brute-forcing a large grid; compute has a carbon cost and, on this data, the marginal
  hyperparameter gain was in the fourth decimal (§4.6).

# 7. Documentation and Code Quality

## 7.1 How this notebook is engineered

* **One configuration object.** Every tunable (paths, sample sizes, CV folds, search budget,
  cost model, seeds) lives in the frozen `Config` dataclass; nothing is hard-coded downstream, so a

* **Leakage discipline.** Split before pruning correlated features; imputation and scaling inside
  the `Pipeline`; the test set scored once per model; hyperparameters tuned by CV on training data

* **Reproducibility.** Fixed `random_state` everywhere (split, folds, searches, forests, simulator),

* **Graceful degradation.** Missing CSVs do not break the notebook: the loader falls back to a

## 7.4 Limitations, risks and next steps

**Limitations of this work — state these before anyone quotes the headline score**

1. **Data provenance.** Results reflect the dataset actually loaded (see `DATA MODE` in §1.3 and in
   the model card). Lab-captured or simulated flows are cleaner and more separable than production
   traffic; treat the metrics as an upper bound, not a production forecast.
2. **Flow-level, context-free decisions.** Each flow is judged alone. Real intrusions are
   *sequences* (scan → exploit → beacon → exfiltrate). Host-level and time-window aggregation
   (flows per host per minute, distinct ports touched, periodicity of beaconing) is the single
   biggest expected accuracy gain — and it is missing here by design, because the given schema is
   one row per flow.
3. **Stationarity assumption.** A random stratified split lets flows from the same attack episode
   fall on both sides of the split, which inflates scores. A **time-based split** (train on days
   1–3, test on days 4–5) is the honest evaluation for a deployed detector and is the recommended
   next experiment.
4. **Rare families remain weak** (§3.5) — the aggregate metric hides them; report per family.
5. **Probabilities are only approximately calibrated** (§5.3); if scores drive a monetary or SLA
   decision, wrap the model in `CalibratedClassifierCV`.
6. **No adversarial evaluation.** Robustness to deliberately shaped traffic was argued (§6.5) but
   not measured; an evasion test suite is required before auto-blocking.
7. **Encrypted traffic** limits what flow statistics can reveal; performance on TLS-heavy networks
   must be re-measured.

**Next steps, in priority order**

| Priority | Action | Expected benefit |
|---|---|---|
| 1 | Time-based (chronological) train/test split | Honest generalisation estimate |
| 2 | Host/window aggregation features | Detects multi-flow campaigns; biggest accuracy lever |
| 3 | Validate on a slice of the customer's real traffic; recalibrate the threshold on their FP budget | Production-realistic performance |
| 4 | Multi-class attack attribution model behind the binary detector | Faster analyst triage and automated playbook selection |
| 5 | Shadow deployment (score, log, never block) for 4 weeks + analyst feedback loop | Real precision measurement without user harm |
| 6 | Gradient-boosting alternatives (XGBoost/LightGBM) + `CalibratedClassifierCV` | Marginal accuracy and better probabilities |
| 7 | Complete the DPIA and publish the model card to customers | Legal and contractual readiness |
